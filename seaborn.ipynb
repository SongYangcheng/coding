## 关系图
关系图是一种用于表示实体及其相互关系的图形化工具。它通常由节点（表示实体）和边（表示实体之间的关系）组成。关系图广泛应用于各种领域，如社交网络分析、知识表示、数据库设计等。
### 散点图-scatterplot
散点图是一种用于显示两个变量之间关系的图表。每个点代表一个数据点，其位置由两个变量的值决定。散点图常用于识别变量之间的相关性、趋势和异常值。
#### sns.scatterplot(data, x, y)相关参数
- x: x轴数据
- y: y轴数据
- hue: 用于区分数据点颜色的变量
- hue_order: 用于设置hue变量的顺序
- palette: 颜色调色板
- size: 用于区分数据点大小的变量，数值越大点越大可简写为s
- sizes: 用于设置数据点大小的范围，格式为(min_size, max_size)
- style: 用于区分数据点样式的变量，映射成不同的点形状
- markers: 用于设置数据点的形状，包括's'（方形）、'o'（圆形）、'^'（三角形）等
- alpha: 数据点的透明度，取值范围为0到1，0表示完全透明，1表示完全不透明
- edgecolor: 数据点边缘的颜色
- legend: 是否显示图例，默认为True, 还可为'scale'或'full','brief', 'scale'表示只显示与大小相关的图例，'full'表示显示所有图例, 'brief'表示显示简要图例
- hue_norm: 用于规范化hue变量的范围
- log_scale: 是否对轴进行对数缩放，可以是布尔值或包含轴名称的列表
- size_norm: 用于规范化size变量的范围,大小归一


#### 多维映射
- X + Y: 基础2维
- X + Y + hue: 3维(颜色)
- X + Y + hue + size: 4维(颜色+大小)
- X + Y + hue + size + style: 5维(颜色+大小+形状)
#### scatterplot()和relpot()的区别
- scatterplot()专门用于绘制散点图，而relplot()是一个更通用的函数，可以绘制散点图和线图，取决于传递的参数。
- relplot()可以创建一个图形网格，允许在多个子图中显示数据，而scatterplot()仅创建单个图形。
- relplot()的默认行为是创建一个散点图，但可以通过设置kind参数为'line'来绘制线图，而scatterplot()始终绘制散点图。

### 折线图-lineplot()函数
lineplot()函数用于创建折线图，显示数据点之间的连续关系。它通过连接数据点来展示趋势和变化，适用于时间序列数据或其他连续变量的数据可视化。
#### sns.lineplot(data, x, y)相关参数
- x: x轴数据
- y: y轴数据
- hue: 用于区分线条颜色的变量
- style: 用于区分线条样式的变量，映射成不同的线条样式
- markers: 用于设置数据点的形状，包括's'（方形）、'o'（圆形）、'^'（三角形）等，当为True时，表示在每个数据点上绘制标记
- markersize: 数据点标记的大小
- markeredgecolor: 数据点标记边缘的颜色
- dashes: 用于设置线条的虚线样式，可以是布尔值或包含线条样式的列表,自动为style变量映射相应的线条样式
- ci: 置信区间的大小，默认为95，可以设置为None表示不显示置信区间
- err_style: 误差条的样式，可以是'band'（带状）或'bars'（误差条）
- errorbar: 用于设置误差条的类型，可以是'sd'（标准差）、'se'（标准误差）或'ci'（置信区间）
- estimator: 用于计算汇总统计量的函数，默认为numpy.mean, 聚合函数中位数可用np.median，也可以传入自定义函数，当x轴对应多个y值时，会对这些y值进行聚合计算
- n_boot: 用于计算置信区间的自助法样本数量，默认为1000
- sort: 是否对x轴数据进行排序，默认为True，会将x轴数据按升序排列
- label: 用于设置线条的标签
- units: 用于指定数据中的重复测量单位，以便正确计算置信区间
#### units和estimator参数区别
|特性|units参数|estimator参数|
|-|-|-|
|作用|指定数据中的重复测量单位|用于计算汇总统计量的函数|
|功能|帮助正确计算置信区间|对y值进行聚合计算|
|使用场景|当数据包含重复测量时使用|当需要汇总统计量时使用|
|数据处理|不改变数据，只影响置信区间计算|对y值进行聚合计算|
|图例|会创建图例|不会创建图例|
|颜色|为每个单位分配颜色|不会影响颜色分配|  

- 使用 hue：当你关心组与组之间的平均差异时。
- 使用 units：当你关心个体与个体之间的差异，或者想看数据中的原始“轨迹”时。
- estimator=None 经常和 units 搭配使用，用来关闭聚合功能，只显示个体线。
#### set_context()函数
set_context()函数用于设置Seaborn图形的上下文环境，以调整图形的整体外观和比例。它允许用户根据不同的展示需求（如论文、演示等）来调整图形的字体大小、线条粗细等参数，从而提高图形的可读性和美观性。

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
contexts = ['paper', 'notebook', 'talk', 'poster']
for i, context in enumerate(contexts):
    sns.set_context(context, font_scale=i) #设置context, font_scale调整字体大小
    plt.subplot(2, 2, i + 1) #创建子图，最后一个参数表示子图位置
    sns.lineplot(x=[1, 2, 3 ,4], y=[1, 4, 2, 3], color='b') #绘制折线图
    plt.title(f'Context: {context}')
plt.tight_layout()
plt.show()
```

### 散点图-relplot()函数
返回的是一个FacetGrid对象，方便进行多图布局调整。<font color='red'>没有办法直接在已有的ax上绘图</font>。
relplot()函数是Seaborn库中的一个高级绘图函数，用于创建关系图。它可以绘制散点图和折线图，取决于传递的参数。relplot()函数支持多维数据映射，通过hue、size和style等参数，可以在同一图形中展示多个变量之间的关系。此外，relplot()还支持创建图形网格，允许在多个子图中显示数据，从而更好地比较不同类别或条件下的数据分布和趋势。
#### sns.relplot(data, x, y, kind='scatter')相关参数
- x: x轴数据
- y: y轴数据
- kind: 图表类型，可以是'scatter'（散点图）或'line'（折线图）按照其形状对应相应的参数参考scatterplot()或lineplot()函数的参数
- hue: 用于区分数据点颜色的变量
- size: 用于区分数据点大小的变量
- style: 用于区分数据点样式的变量
- col: 用于创建列子图的变量,分为多列显示
- row: 用于创建行子图的变量,分为多行显示
- col_wrap: 用于设置列子图的换行数量,假如col有12列且col_wrap=4,则会显示3行4列的子图
- row_order: 用于设置行子图的顺序
- col_order: 用于设置列子图的顺序
- aspect: 用于设置子图的宽高比为宽度/高度
- height: 用于设置每个子图的高度，单位为英寸
- alpha: 数据点的透明度，取值范围为0到1，0表示完全透明，1表示完全不透明  
⚠️line特有参数:
   - estimator: 聚合函数
     * 'mean'(默认)
     * 'median', 'sum'等
   - errorbar: 误差表示
     * 'se': 标准误(默认)
     * 'sd': 标准差
     * 'ci': 置信区间
     * ('pi', 95): 预测区间
   - markers: True/False，是否添加标记点
   - dashes: True/False，是否使用虚线
   - sort: True/False，是否排序X轴
   - err_style: 误差带样式('band'/'bars')

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']  #设置中文字体为SimHei
tips = pd.read_csv(r'D:\python_demo\coding\data\seaborndata\tips.csv')
sns.relplot(
    data=tips, #指定数据集
    x='消费金额 ($)', #指定x轴变量
    y='小费金额 ($)', #指定y轴变量
    # col='就餐时间', #指定列变量
    hue='是否吸烟', #指定颜色变量
    style='客人性别', #指定样式变量
    palette=['red', 'blue'], #指定调色板
    # markers=['o', 's'], #指定点的样式
    s = 200, #指定点的大小
    kind='scatter', #指定关系图类型
    color='green', #指定点颜色
)
```

#### relplot()函数和scatterplot()函数的区别
relplot()函数是Seaborn库中的一个高级绘图函数，用于创建关系图（relational plots）。它可以用于绘制散点图（scatter plots）和线图（line plots），并且支持分面（faceting）功能，可以根据一个或多个分类变量将数据分成多个子图。
scatterplot()函数是Seaborn库中的一个基础绘图函数，专门用于创建散点图。它不支持分面功能，主要用于绘制单个散点图。
relplot()可以是scatterplot()和lineplot()的高级接口，它可以根据kind参数的不同调用相应的绘图函数，从而实现更灵活的绘图需求。

### 核密度图-kdeplot()函数
kdeplot()函数用于创建核密度估计图，显示数据的概率密度分布。它通过平滑数据点来估计数据的分布情况，适用于连续变量的数据可视化。
#### sns.kdeplot(data, x, y)相关参数
- clip: 用于指定绘图的范围，可以是一个元组，包含两个值，分别表示x轴的最小值和最大值
- fill: 是否填充密度曲线下方的区域，默认为False，表示不填充
- bw_adjust: 用于调整核密度估计的带宽。它是一个乘数，用于调整默认带宽的大小。较大的bw_adjust值会增加带宽，从而使密度估计更平滑；较小的bw_adjust值会减小带宽，从而使密度估计更尖锐。
- bw_method: 用于指定核密度估计的带宽选择方法。可选的值包括'scott'（Scott's Rule）、'silverman'（Silverman's Rule）和'normal_reference'（Normal Reference Rule）。
- gridsize: 用于指定用于计算核密度估计的网格点数量
- cut: 用于指定在数据范围之外绘制密度估计的扩展范围。它是一个正数，表示在数据范围之外绘制密度估计的倍数
- levels: 用于指定等高线的数量或具体值。它可以是一个整数，表示等高线的数量；也可以是一个序列，表示具体的等高线值。
- thresh: 用于指定绘制密度估计的阈值。它是一个非负数，表示只有当密度估计值大于该阈值时，才会绘制密度估计。
- cumulative: 是否绘制累积密度估计，默认为False，表示绘制普通密度估计
- complementary: 是否绘制补充密度估计，默认为False，表示不绘制补充密度估计
- common_norm: 是否对多个数据集进行归一化处理，默认为True，表示对所有数据集进行归一化
- common_grid: 是否在多个数据集上使用相同的网格点，默认为True，表示使用相同的网格点
- log_scale: 是否对轴进行对数缩放，可以是布尔值或包含轴名称的列表
- multiple: 用于指定密度估计的绘制方式。可选的值包括'layer'（层叠）、'dodge'（错开）和'stack'（堆叠）, 'fill'（填充）。

#### cut参数
cut参数用于控制核密度估计的范围。它指定了在数据范围之外绘制密度估计的距离。默认情况下，cut参数的值为3，这意味着密度估计将在数据范围之外延伸3个标准差。如果将cut参数设置为0，则密度估计将仅在数据范围内绘制。


```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
penguins = pd.read_csv(r'D:\python_demo\coding\data\seaborndata\penguins.csv')
print(penguins.columns.tolist())
plt.rcParams['font.family'] = ['SimHei', 'Times New Roman']  #设置中文字体为SimHei
fig, axes = plt.subplots(2, 2, figsize=(12, 5))
sns.kdeplot( # 创建一个核密度图
    data=penguins,
    x='鳍长 (毫米)',
    ax=axes[0, 0],
    color='red',
    cut=3
)
sns.kdeplot(
    data=penguins,
    x='鳍长 (毫米)',
    cut=1,
    color='red',
    ax=axes[0, 1]
)
sns.kdeplot( # 创建一个核密度图
    data=penguins,
    x='鳍长 (毫米)',
    ax=axes[1, 0],
    cut=2,
    color='red',
)
sns.kdeplot( # 创建一个核密度图
    data=penguins,
    x='鳍长 (毫米)',
    ax=axes[1, 1],
    color='red',
    cut=3
)
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
penguins = pd.read_csv(r'D:\python_demo\coding\data\seaborndata\penguins.csv')

df = sns.load_dataset('fmri').copy()

df = df.rename(
    columns={
        'subject': '受试者',
        'timepoint': '时间点',
        'event': '事件类型',
        'region': '大脑区域',
        'signal': '信号强度'
    }
)

区域顺序 = ['parietal', 'frontal']
事件顺序 = ['stim', 'cue']
df_mean = df.groupby(['时间点', '事件类型', '大脑区域'], as_index=False).agg(信号强度=('信号强度', 'mean'))

fig, axes = plt.subplots(2, 2, figsize=(12, 5), constrained_layout=True)
axes = axes.ravel() #将二维数组展平为一维数组，方便迭代
sns.lineplot(
    data=df_mean.query("时间点 == 'stim' and 大脑区域 == 'parietal'"), #筛选数据
    ax=axes[0],
    x='时间点',
    y='信号强度',
    linewidth=2,
    linestyle='-',
    markers='o',
    alpha=0.9,
    estimator=None, #聚合函数，表示不聚合
    errorbar=None, #不显示误差条
)
axes[0].set_title('A. 事件点=stim,大脑区域=parietal')
axes[0].set_xlabel('时间点')
axes[0].set_ylabel('信号强度')

sns.lineplot(
    data=df_mean.query("大脑区域 == 'parietal'"), #筛选数据
    ax=axes[1],
    x='时间点',
    y='信号强度',
    hue='事件类型',
    hue_order=事件顺序,
    palette='Set1',
    estimator=np.mean, #聚合函数，表示计算均值
    errorbar=('ci', 95), #显示95%置信区间误差条
    linewidth=2,
    legend='brief'
)
axes[1].set_title('B. 大脑区域=parietal')
axes[1].set_xlabel('时间点')
axes[1].set_ylabel('信号强度')

sns.lineplot(
    data=df_mean.query("时间点 == 'stim'"), #筛选数据
    ax=axes[2],
    x='时间点',
    y='信号强度',
    hue='大脑区域',
    hue_order=区域顺序,
    palette='Set2',
    estimator=np.mean, #聚合函数，表示计算均值
    # errorbar=('ci', 95), #显示95%置信区间误差条
    linewidth=2,
    dashes=True,
    legend='full'
)
axes[2].set_title('大脑区域=stim')
axes[2].set_xlabel('时间点')
axes[2].set_ylabel('信号强度')

sns.lineplot(
    data=df,
    ax=axes[3],
    x='时间点',
    y='信号强度',
    hue='事件类型',
    hue_order=事件顺序,
    palette='tab10',
    style='大脑区域',
    style_order=区域顺序,
    markers=True,
    dashes=True, #启动线型：虚线
    estimator=np.mean,
    errorbar='sd', #显示95%置信区间误差条
    err_style='bars',
    err_kws={
        'capsize': 5,
        'capthick': 1,
        'elinewidth': 1
    },
    linewidth=2,
    alpha=0.9,
)
```

## 分类图
分类图是一种用于显示分类变量与数值变量之间关系的图表。它通过将数据分组并计算每个类别的统计量（如均值、中位数等）来展示不同类别之间的差异。分类图常用于比较不同组别的数据分布和趋势，帮助识别类别之间的关系和模式。
### 条形图-barplot()函数
barplot()函数用于创建条形图，显示分类变量与数值变量之间的关系。它通过计算每个类别的均值（或其他聚合统计量）来绘制条形图，并可以显示误差条以表示数据的变异性。

#### 参数说明
统计参数  
- estimator：用于计算每个x值对应的y值的函数。默认情况下，estimator=None，表示使用所有数据点进行估计。可以将其设置为numpy.mean、numpy.median等函数，以计算每个x值对应的y值的平均值、中位数等。
- errorbar：用于**指定误差条**⚠️的类型。可以设置为'sd'（标准差）、'ci'（置信区间）等。默认情况下，errorbar=None，表示不绘制误差条。
- capsize：用于指定误差条的帽子大小。默认情况下，capsize=None，表示不绘制帽子。可以将其设置为一个正数，以指定帽子的大小。
- order：用于指定数据点的绘制顺序。默认情况下，order=None，表示按照数据的顺序绘制。可以将其设置为一个整数列表，以指定数据点的绘制顺序。  
- hue: 用于指定分组变量。可以将数据按照hue变量的不同取值进行分组，并为每个组绘制不同的颜色。默认情况下，hue=None，表示不进行分组。
样式参数
- palette：用于指定颜色调色板。可以设置为预定义的调色板名称（如'deep'、'muted'、'bright'等），也可以设置为自定义的颜色列表。默认情况下，palette=None，表示使用默认的颜色调色板。
- color：用于指定单一颜色。可以设置为颜色名称（如'red'、'blue'等）、十六进制颜色代码（如'#FF0000'）等。默认情况下，color=None，表示使用默认颜色。  
布局参数
- dodge：用于指定是否对分组数据进行错开显示。默认情况下，dodge=False，表示不进行错开显示。可以将其设置为True，以对分组数据进行错开显示。  
- legend：用于指定是否显示图例。默认情况下，legend='auto'，表示根据需要显示图例。可以将其设置为True或False，以强制显示或隐藏图例。
- ax：用于指定绘图的坐标轴对象。默认情况下，ax=None，表示在当前坐标轴上绘图。可以将其设置为一个matplotlib的Axes对象，以在指定的坐标轴上绘图。  
  ⚠️ 使用palette时必须指定hue参数

### countplot()函数
countplot()函数用于创建计数条形图，显示每个类别的<font color='red'>**频数**</font>。它通过计算每个类别的出现次数来绘制条形图，适用于分类变量的数据可视化。
#### 参数说明
- x: 用于指定分类变量的列名或数组。默认情况下，x=None，表示不使用x轴数据。

### 箱线图绘制-boxplot()函数
箱线图（Box Plot）是一种用于显示数据分布情况的统计图表。它通过五个统计量来描述数据的分布：最小值、第一四分位数（Q1）、中位数（Q2）、第三四分位数（Q3）和最大值。箱线图还可以显示异常值（outliers），即远离其他数据点的值。   
上边缘：Q3+1.5IQR  
上四分位数(Q3)  
中位数：Q2  
下四分位数(Q1)  
下边缘：Q1-1.5IQR  
IQR(四分位距)=Q3-Q1
##### boxplot()函数参数
- x, y, hue：用于指定分类变量和数值变量。x和y分别表示分类变量和数值变量，hue用于指定分组变量。
- order, hue_order：用于指定分类变量和分组变量的绘制顺序。
- palette：用于指定颜色调色板。
- width：用于指定箱线图的宽度。
- showfliers：用于指定是否显示异常值, 离群点类似气泡圆圈状在图形中。
- showmeans：用于指定是否显示均值。
- notch: 用于指定是否绘制缺口箱线图。
- orient：用于指定箱线图的方向。
- whis=[min, max]: 用于指定须的范围, min和max分别表示须的下限和上限。
- whiskerprops: 用于自定义须的样式属性。
- 样式参数(props系列):
   - boxprops: 箱体样式字典
   - whiskerprops: 箱须样式字典
   - capprops: 箱须端点样式字典
   - medianprops: 中位线样式字典
   - meanprops: 均值点样式字典
   - flierprops: 异常值样式字典
- fill：用于指定是否填充箱体颜色。
- dodge：用于指定是否对分组数据进行错开显示。 
- palette：用于指定颜色调色板。可选的值包括预定义的调色板名称（如'deep'、'muted'、'bright'等）或自定义的颜色列表。     
常用技巧：  
boxplot + swarmplot:不重叠数据点展示分布， 箱型图和蜂群图结合 
boxplot + stripplot:展示数据分布和具体数据点， 箱型图和抖动散点图结合
### stripplot()函数
stripplot()函数用于创建抖动散点图（strip plot），显示分类变量与数值变量之间的关系。它通过在数据点周围添加随机抖动来避免重叠，从而展示数据的分布情况。stripplot()函数适用于较大的数据集，因为它不会根据数据点的数量调整位置，而是通过添加抖动来减少重叠。
### swarmplot()函数参数
- x, y, hue：用于指定分类变量和数值变量。x和y分别表示分类变量和数值变量，hue用于指定分组变量。
- order, hue_order：用于指定分类变量和分组变量的绘制顺序
- palette：用于指定颜色调色板。
- size：用于指定数据点的大小。
- marker：用于指定数据点的形状。
- dodge：用于指定是否对分组数据进行错开显示。
- edgecolor：用于指定数据点的边缘颜色。
- linewidth：用于指定数据点的边缘宽度。
- alpha：用于指定数据点的透明度。
- orient：用于指定图形的方向。
- fill：用于指定是否填充数据点颜色。
- saturation：用于指定颜色的饱和度。
- legend：用于指定是否显示图例。
- jitter：用于指定抖动的程度。
- ax：用于指定绘图的坐标轴对象。

##### swarmplot()函数和stripplot()函数区别
swarmplot()函数用于创建蜂群图（swarm plot），它通过调整数据点的位置来避免重叠，从而更清晰地展示数据的分布情况。swarmplot()函数适用于较小的数据集，因为它会根据数据点的数量自动调整位置，以确保每个数据点都能被清晰地看到。
stripplot()函数用于创建抖动散点图（strip plot），它通过在数据点周围添加随机抖动来避免重叠，从而展示数据的分布情况。stripplot()函数适用于较大的数据集，因为它不会根据数据点的数量调整位置，而是通过添加抖动来减少重叠。

### 小提琴图-violinplot()函数
violinplot 的“琴身”是一个核密度估计 (KDE) 图。
小提琴图（Violin Plot）是一种结合了箱线图和核密度估计图的统计图表。它通过显示数据的分布情况和概率密度来提供更丰富的信息。小提琴图的形状类似于小提琴，因此得名。
#### violinplot()函数参数
- x, y, hue：用于指定分类变量和数值变量。x和y分别表示分类变量和数值变量，hue用于指定分组变量。
- order, hue_order：用于指定分类变量和分组变量的绘制顺序
- palette：用于指定颜色调色板。
- width：用于指定小提琴图的宽度。
- inner：用于指定小提琴图内部的显示方式。可选的值包括'box'（显示箱线图）、'quartile'（显示四分位数）、'point'（显示数据点）和'stick'（显示数据点的垂直线）。
- cut：用于指定核密度估计的范围。默认情况下，cut=2，会在数据范围之外延伸2个标准差。可以将其设置为0，以仅在数据范围内绘制密度估计。用于控制小提琴的尾巴长度。
- scale：用于指定小提琴图的缩放方式。可选的值包括'area'（根据面积缩放）、'count'（根据计数缩放）和'width'（根据宽度缩放）。
- bw：用于指定核密度估计的带宽。带宽控制了密度估计的平滑程度。较小的带宽会导致更尖锐的密度估计，而较大的带宽会导致更平滑的密度估计。
- bw_adjust：用于调整核密度估计的带宽。它是一个乘数，用于调整默认带宽的大小。较大的bw_adjust值会增加带宽，从而使密度估计更平滑；较小的bw_adjust值会减小带宽，从而使密度估计更尖锐。。
- gridsize：用于指定核密度估计的网格大小。它决定了用于计算密度估计的点的数量。较大的gridsize值会导致更精细的密度估计，而较小的gridsize值会导致更粗糙的密度估计。
-  split：用于指定是否将小提琴图分割为两半。默认情况下，split=False，表示不分割小提琴图。可以将其设置为True，以将小提琴图分割为两半，分别显示hue变量的不同取值。
- ⚠️⚠️⚠️density_norm 是在 histplot (直方图) 或 kdeplot (核密度图) 中用来控制“密度归一化”（即Y轴显示计数还是概率密度）的参数。
#### 常用组合
boxplot + violinplot: 展示数据分布的多样性， 箱型图和小提琴图结合
violinplot + swarmplot: 展示数据分布和具体数据点， 小提琴图和蜂群图结合
violinplot + stripplot: 展示数据分布和具体数据点， 小提琴图和抖动散点图结合

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# 加载多个数据集
penguins = sns.load_dataset('penguins').dropna()
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')

print(f"Penguins数据集形状: {penguins.shape}")
print(f"Tips数据集形状: {tips.shape}")
print(f"Iris数据集形状: {iris.shape}")
print(tips.columns.tolist())
print(iris.columns.tolist())
# ============================================================
# 画布 1: 基础用法和核心参数
# ============================================================
fig1, axes1 = plt.subplots(2, 5, figsize=(24, 10))  # 创建2行5列子图布局，总画布大小24x10英寸
fig1.suptitle('Seaborn Violinplot 教程 - 画布1: 基础用法与核心参数',  # 设置整个画布的总标题
              fontsize=16,  # 总标题字体大小16
              fontweight='bold',  # 总标题加粗显示
              y=0.995)  # 总标题垂直位置略低于顶部（避免裁剪）

# ========== 子图 A：最基础的小提琴图 ==========
ax = axes1[0, 0]  # 获取第1行第1列子图对象
sns.violinplot(data=penguins,  # 数据源：企鹅数据集
               x='species',  # x轴：企鹅物种（分类变量）
               y='body_mass_g',  # y轴：体重（数值变量）
               ax=ax,  # 指定绘制到当前子图
               color='steelblue')  # 设置单一颜色为钢蓝色
ax.set_title('A. 基础小提琴图\n密度+分布', fontsize=11, fontweight='bold')  # 子图标题说明密度分布功能
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签：企鹅种类
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签：体重（克）
ax.text(0.02,  # 文本水平位置：左边距2%
        0.98,  # 文本垂直位置：顶部98%
        '小提琴图特点:\n显示密度分布\n对称的核密度估计\n宽度=数据密度',  # 详细说明小提琴图核心特点
        transform=ax.transAxes,  # 使用相对坐标系
        fontsize=8,  # 字体大小8
        verticalalignment='top',  # 顶部对齐
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))  # 圆角浅黄色半透明背景框

# ========== 子图 B：添加hue分组 ==========
ax = axes1[0, 1]  # 获取第1行第2列子图对象
sns.violinplot(data=penguins,  # 数据源
               x='species',  # x轴：物种
               y='body_mass_g',  # y轴：体重
               hue='sex',  # hue参数：按性别分组，并排显示
               ax=ax)  # 绘制到当前子图，自动配色
ax.set_title('B. 分组小提琴图\nhue分组', fontsize=11, fontweight='bold')  # 标题说明分组功能
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签
ax.legend(title='性别', loc='upper left', fontsize=8)  # 图例：标题"性别"，左上角，字体8
ax.text(0.02,  # 文本位置
        0.98,
        '参数: hue="sex"\n并排显示不同组',  # 说明hue参数并排显示效果
        transform=ax.transAxes,
        fontsize=9,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))  # 浅蓝色背景框

# ========== 子图 C：split分割显示 ==========
ax = axes1[0, 2]  # 获取第1行第3列子图对象
sns.violinplot(data=penguins,  # 数据源
               x='species',  # x轴：物种
               y='body_mass_g',  # y轴：体重
               hue='sex',  # hue：按性别分组
               split=True,  # split=True：左右分割显示（仅限2组hue）
               ax=ax)  # 绘制到当前子图，自动配色
ax.set_title('C. 分割小提琴图\nsplit=True', fontsize=11, fontweight='bold')  # 标题说明分割功能
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签
ax.legend(title='性别', loc='upper left', fontsize=8)  # 图例设置
ax.text(0.02,  # 文本位置
        0.98,
        'split=True:\n左右分割显示\n更紧凑的对比\n仅适用于2组hue',  # 说明split适用条件和优势
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))  # 浅黄色背景框

# ========== 子图 D：横向小提琴图 ==========
ax = axes1[0, 3]  # 获取第1行第4列子图对象
sns.violinplot(
               data=tips,  # 数据源：tips小费数据集
               x='total_bill',  # x轴：账单总额（数值）
               y='day',  # y轴：星期（分类）-交换实现横向
               hue='day',  # hue：按星期着色
               palette='viridis',  # viridis连续色彩调色板
               legend=False,  # 隐藏图例
               ax=ax)  # 绘制到当前子图
ax.set_title('D. 横向小提琴图\nTips数据集', fontsize=11, fontweight='bold')  # 标题说明横向和数据集
ax.set_xlabel('账单金额 ($)', fontsize=10)  # x轴标签：美元
ax.set_ylabel('星期', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        '参数: x和y互换\n使用tips数据集\n展示消费分布',  # 说明横向参数和应用场景
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))  # 浅绿色背景框

# ========== 子图 E：inner参数 - box ==========
ax = axes1[0, 4]  # 获取第1行第5列子图对象
sns.violinplot(data=penguins,  # 数据源
               x='species',  # x轴：物种
               y='flipper_length_mm',  # y轴：鳍长
               hue='species',  # 分组：按物种着色
               palette='Set2',  # Set2柔和色彩调色板
               legend=False,  # 隐藏图例
               inner='box',  # inner="box"：内部显示微型箱线图
               ax=ax)  # 绘制目标
ax.set_title('E. 内部显示箱线图\ninner="box"', fontsize=11, fontweight='bold')  # 标题说明内部箱线
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.set_ylabel('鳍长 (mm)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'inner="box":\n内部显示微型箱线图\n结合两种图优点',  # 说明箱线+密度组合优势
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lavender', alpha=0.5))  # 淡紫色背景框

# ========== 子图 F：inner参数 - quartile ==========
ax = axes1[1, 0]  # 获取第2行第1列子图对象
sns.violinplot(data=tips,  # 数据源：tips数据集
               x='day',  # x轴：星期
               y='tip',  # y轴：小费
               hue='day',  # 分组：按星期
               palette='coolwarm',  # 冷暖色系调色板
               legend=False,  # 隐藏图例
               inner='quartile',  # inner="quartile"：显示Q1、Q2、Q3四分位线
               ax=ax)  # 绘制目标
ax.set_title('F. 显示四分位数\ninner="quartile"', fontsize=11, fontweight='bold')  # 标题说明四分位显示
ax.set_xlabel('星期', fontsize=10)  # x轴标签
ax.set_ylabel('小费 ($)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'inner="quartile":\n显示四分位线\n标记Q1,Q2,Q3',  # 说明四分位线标记
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='mistyrose', alpha=0.5))  # 玫瑰色背景框

# ========== 子图 G：inner参数 - point ==========
ax = axes1[1, 1]  # 获取第2行第2列子图对象
sns.violinplot(data=penguins,  # 数据源
               x='island',  # x轴：岛屿
               y='bill_length_mm',  # y轴：喙长
               hue='island',  # 分组：按岛屿
               palette='muted',  # 柔和色彩调色板
               legend=False,  # 隐藏图例
               inner='point',  # inner="point"：显示每个原始数据点
               ax=ax)  # 绘制目标
ax.set_title('G. 显示单个点\ninner="point"', fontsize=11, fontweight='bold')  # 标题说明数据点显示
ax.set_xlabel('岛屿', fontsize=10)  # x轴标签
ax.set_ylabel('喙长 (mm)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'inner="point":\n显示每个数据点\n适合小数据集',  # 说明适用场景
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightcyan', alpha=0.5))  # 浅青色背景框

# ========== 子图 H：inner参数 - stick ==========
ax = axes1[1, 2]  # 获取第2行第3列子图对象
sns.violinplot(data=iris,  # 数据源：iris鸢尾花数据集
               x='species',  # x轴：花种
               y='sepal_length',  # y轴：萼片长度
               hue='species',  # 分组：按花种
               palette='Spectral',  # 光谱色调色板
               legend=False,  # 隐藏图例
               inner='stick',  # inner="stick"：每条竖线代表1个数据点
               ax=ax)  # 绘制目标
ax.set_title('H. 显示棒状标记\ninner="stick" (Iris)', fontsize=11, fontweight='bold')  # 标题说明棒状标记
ax.set_xlabel('鸢尾花种类', fontsize=10)  # x轴标签
ax.set_ylabel('萼片长度 (cm)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'inner="stick":\n显示竖线标记\n每条线=1个数据',  # 说明棒状标记含义
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='peachpuff', alpha=0.5))  # 桃色背景框

# ========== 子图 I：inner参数 - None ==========
ax = axes1[1, 3]  # 获取第2行第4列子图对象
sns.violinplot(data=penguins,  # 数据源
               x='species',  # x轴：物种
               y='body_mass_g',  # y轴：体重
               hue='species',  # 分组
               palette='tab10',  # 10色分类调色板
               legend=False,  # 隐藏图例
               inner=None,  # inner=None：不显示任何内部标记
               ax=ax)  # 绘制目标
ax.set_title('I. 无内部标记\ninner=None', fontsize=11, fontweight='bold')  # 标题说明纯密度显示
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'inner=None:\n只显示密度曲线\n最简洁的形式',  # 说明最简洁形式
        transform=ax.transAxes,
        fontsize=9,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))  # 浅黄色背景框

# ========== 子图 J：调整带宽 bw_adjust ==========
ax = axes1[1, 4]  # 获取第2行第5列子图对象
sns.violinplot(data=tips,  # 数据源：tips数据集
               x='time',  # x轴：时间段（午餐/晚餐）
               y='total_bill',  # y轴：账单总额
               hue='time',  # 分组：按时间段
               palette='rocket',  # rocket红黑渐变调色板
               legend=False,  # 隐藏图例
               bw_adjust=0.5,  # bw_adjust=0.5：带宽调整为默认的50%（更细节）
               ax=ax)  # 绘制目标
ax.set_title('J. 调整平滑度\nbw_adjust=0.5', fontsize=11, fontweight='bold')  # 标题说明平滑度调整
ax.set_xlabel('时间段', fontsize=10)  # x轴标签
ax.set_ylabel('账单金额 ($)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'bw_adjust:\n<1: 更多细节\n>1: 更平滑\n默认1.0',  # 详细说明带宽调整效果
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))  # 浅蓝色背景框

plt.tight_layout()  # 自动调整子图间距，避免重叠
plt.show()  # 显示画布1的10个子图
```

### 抖动散点图-stripplot()函数
stripplot()函数用于创建抖动散点图（strip plot），显示分类变量与数值变量之间的关系。它通过在数据点周围添加随机抖动来避免重叠，从而展示数据的分布情况。stripplot()函数适用于较大的数据集，因为它不会根据数据点的数量调整位置，而是通过添加抖动来减少重叠。
#### stripplot()函数参数
- x, y, hue：用于指定分类变量和数值变量。x和y
- 分别表示分类变量和数值变量，hue用于指定分组变量。
- order, hue_order：用于指定分类变量和分组变量的绘制顺序
- palette：用于指定颜色调色板。
- size：用于指定数据点的大小。
- marker：用于指定数据点的形状。
- dodge：用于指定是否对分组数据进行错开显示, 参数为True或False为分开和不分开。
- jitter：用于指定抖动的程度。
- alpha：用于指定数据点的透明度。
- ax：用于指定绘图的坐标轴对象。
- legend：用于指定是否显示图例。
- color：用于指定单一颜色。
- orient：用于指定图形的方向（“v”表示垂直方向，“h”表示水平方向”）。
- edgecolor：用于指定数据点的边缘颜色。
- ⚠️⚠️⚠️无split：用于指定是否将数据点分割为两半。默认情况下，split=False，表示不分割数据点。可以将其设置为True，以将数据点分割为两半，分别显示hue变量的不同取值。

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
plt.rcParams['font.family'] = ['SimHei', 'Times New Roman']  # 设置字体以支持中文显示（Mac系统）
# 加载多个数据集
penguins = sns.load_dataset('penguins').dropna()  # 加载企鹅数据集并删除缺失值
tips = sns.load_dataset('tips')  # 加载tips小费数据集
iris = sns.load_dataset('iris')  # 加载iris鸢尾花数据集

print(f"Penguins数据集形状: {penguins.shape}")  # 输出企鹅数据集行数×列数
print(f"Tips数据集形状: {tips.shape}")  # 输出tips数据集行数×列数
print(f"\nStripplot特点: 显示每个原始数据点")  # 说明stripplot核心特点
print("适合小到中等规模的数据集")  # 说明适用数据规模

# ============================================================
# 画布 1: 基础用法和核心参数
# ============================================================
fig1, axes1 = plt.subplots(2, 5, figsize=(24, 10))  # 创建2行5列子图布局，总画布大小24x10英寸
fig1.suptitle('Seaborn Stripplot 教程 - 画布1: 基础用法与核心参数',  # 设置整个画布的总标题
              fontsize=16,  # 总标题字体大小16
              fontweight='bold',  # 总标题加粗显示
              y=0.995)  # 总标题垂直位置略低于顶部（避免裁剪）

# ========== 子图 A：最基础的散点条图 ==========
ax = axes1[0, 0]  # 获取第1行第1列子图对象
sns.stripplot(data=penguins,  # 数据源：企鹅数据集
              x='species',  # x轴：企鹅物种（分类变量）
              y='body_mass_g',  # y轴：体重（数值变量）
              ax=ax,  # 指定绘制到当前子图
              color='steelblue')  # 设置单一颜色为钢蓝色
ax.set_title('A. 基础散点条图\n显示所有数据点', fontsize=11, fontweight='bold')  # 子图标题说明显示所有数据
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签：企鹅种类
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签：体重（克）
ax.text(0.02,  # 文本水平位置：左边距2%
        0.98,  # 文本垂直位置：顶部98%
        'Stripplot特点:\n 显示每个数据点\n保留原始信息\n 点可能重叠',  # 详细说明stripplot核心特点
        transform=ax.transAxes,  # 使用相对坐标系
        fontsize=8,  # 字体大小8
        verticalalignment='top',  # 顶部对齐
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))  # 圆角浅黄色半透明背景框

# ========== 子图 B：添加hue分组 ==========
ax = axes1[0, 1]  # 获取第1行第2列子图对象
sns.stripplot(data=penguins,  # 数据源
              x='species',  # x轴：物种
              y='body_mass_g',  # y轴：体重
              hue='sex',  # hue参数：按性别分组，不同颜色
              ax=ax)  # 绘制到当前子图，自动配色
ax.set_title('B. 分组散点条图\nhue分组', fontsize=11, fontweight='bold')  # 标题说明分组功能
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签
ax.legend(title='性别', loc='upper left', fontsize=8)  # 图例：标题"性别"，左上角，字体8
ax.text(0.02,  # 文本位置
        0.98,
        '参数: hue="sex"\n不同颜色区分组',  # 说明hue参数颜色分组效果
        transform=ax.transAxes,
        fontsize=9,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))  # 浅蓝色背景框

# ========== 子图 C：dodge参数分开显示 ==========
ax = axes1[0, 2]  # 获取第1行第3列子图对象
sns.stripplot(data=penguins,  # 数据源
              x='species',  # x轴：物种
              y='body_mass_g',  # y轴：体重
              hue='sex',  # hue：按性别分组
              dodge=True,  # dodge=True：不同hue组水平并排显示
              ax=ax)  # 绘制到当前子图，自动配色
ax.set_title('C. 分组并排显示\ndodge=True', fontsize=11, fontweight='bold')  # 标题说明并排效果
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签
ax.legend(title='性别', loc='upper left', fontsize=8)  # 图例设置
ax.text(0.02,  # 文本位置
        0.98,
        'dodge=True:\n不同hue组并排\n更清晰的对比',  # 说明dodge并排优势
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))  # 浅黄色背景框

# ========== 子图 D：jitter参数控制抖动 ==========
ax = axes1[0, 3]  # 获取第1行第4列子图对象
sns.stripplot(data=tips,  # 数据源：tips小费数据集
              x='day',  # x轴：星期
              y='total_bill',  # y轴：账单总额
              hue='day',  # 分组：按星期着色
              palette='viridis',  # viridis连续色彩调色板
              legend=False,  # 隐藏图例
              jitter=0.3,  # jitter=0.3：水平抖动范围0.3（0-1之间）
              ax=ax)  # 绘制目标
ax.set_title('D. 调整抖动范围\njitter=0.3', fontsize=11, fontweight='bold')  # 标题说明抖动控制
ax.set_xlabel('星期', fontsize=10)  # x轴标签
ax.set_ylabel('账单金额 ($)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'jitter参数:\nTrue/False/数值\n控制水平抖动\n避免点重叠',  # 说明jitter三种用法
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))  # 浅绿色背景框

# ========== 子图 E：jitter=False无抖动 ==========
ax = axes1[0, 4]  # 获取第1行第5列子图对象
sns.stripplot(data=penguins,  # 数据源
              x='island',  # x轴：岛屿
              y='flipper_length_mm',  # y轴：鳍长
              hue='island',  # 分组：按岛屿
              palette='Set2',  # Set2柔和色彩调色板
              legend=False,  # 隐藏图例
              jitter=False,  # jitter=False：无水平抖动，点垂直对齐
              ax=ax)  # 绘制目标
ax.set_title('E. 无抖动对齐\njitter=False', fontsize=11, fontweight='bold')  # 标题说明无抖动效果
ax.set_xlabel('岛屿', fontsize=10)  # x轴标签
ax.set_ylabel('鳍长 (mm)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'jitter=False:\n点垂直对齐\n可能大量重叠\n适合离散数据',  # 说明无抖动特点和适用场景
        transform=ax.transAxes,
        fontsize=7,  # 字体小（内容多）
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lavender', alpha=0.5))  # 淡紫色背景框

# ========== 子图 F：调整点的大小 ==========
ax = axes1[1, 0]  # 获取第2行第1列子图对象
sns.stripplot(data=tips,  # 数据源
              x='time',  # x轴：时间段（午餐/晚餐）
              y='tip',  # y轴：小费
              hue='time',  # 分组：按时间段
              palette='coolwarm',  # 冷暖色系调色板
              legend=False,  # 隐藏图例
              size=8,  # size=8：点大小8点（默认5）
              ax=ax)  # 绘制目标
ax.set_title('F. 调整点大小\nsize=8', fontsize=11, fontweight='bold')  # 标题说明点大小调整
ax.set_xlabel('时间段', fontsize=10)  # x轴标签
ax.set_ylabel('小费 ($)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'size参数:\n控制点的大小\n默认5\n更大更醒目',  # 说明size参数效果
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='mistyrose', alpha=0.5))  # 玫瑰色背景框

# ========== 子图 G：调整透明度 ==========
ax = axes1[1, 1]  # 获取第2行第2列子图对象
sns.stripplot(data=penguins,  # 数据源
              x='species',  # x轴：物种
              y='bill_length_mm',  # y轴：喙长
              hue='species',  # 分组
              palette='muted',  # 柔和色彩调色板
              legend=False,  # 隐藏图例
              alpha=0.4,  # alpha=0.4：点透明度40%
              ax=ax)  # 绘制目标
ax.set_title('G. 调整透明度\nalpha=0.4', fontsize=11, fontweight='bold')  # 标题说明透明度调整
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.set_ylabel('喙长 (mm)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'alpha参数:\n0-1之间\n降低透明度\n看清重叠区域',  # 说明alpha参数作用
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightcyan', alpha=0.5))  # 浅青色背景框

# ========== 子图 H：自定义点样式marker ==========
ax = axes1[1, 2]  # 获取第2行第3列子图对象
sns.stripplot(data=iris,  # 数据源：iris鸢尾花数据集
              x='species',  # x轴：花种
              y='sepal_length',  # y轴：萼片长度
              hue='species',  # 分组：按花种
              palette='Spectral',  # 光谱色调色板
              legend=False,  # 隐藏图例
              marker='D',  # marker='D'：菱形标记（默认'o'圆形）
              size=6,  # 点大小：6点
              ax=ax)  # 绘制目标
ax.set_title('H. 自定义点形状\nmarker="D" (Iris)', fontsize=11, fontweight='bold')  # 标题说明自定义形状
ax.set_xlabel('鸢尾花种类', fontsize=10)  # x轴标签
ax.set_ylabel('萼片长度 (cm)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'marker参数:\no圆形(默认)\nD菱形\ns正方形\n^三角形等',  # 列出常见marker形状
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='peachpuff', alpha=0.5))  # 桃色背景框

# ========== 子图 I：添加边框 ==========
ax = axes1[1, 3]  # 获取第2行第4列子图对象
sns.stripplot(data=penguins,  # 数据源
              x='island',  # x轴：岛屿
              y='body_mass_g',  # y轴：体重
              hue='island',  # 分组
              palette='tab10',  # 10色分类调色板
              legend=False,  # 隐藏图例
              edgecolor='black',  # edgecolor='black'：点边框颜色黑色
              linewidth=1,  # linewidth=1：边框线宽1点
              ax=ax)  # 绘制目标
ax.set_title('I. 添加点边框\nedgecolor + linewidth', fontsize=11, fontweight='bold')  # 标题说明边框设置
ax.set_xlabel('岛屿', fontsize=10)  # x轴标签
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        '边框样式:\nedgecolor边框色\nlinewidth边框宽\n增强视觉效果',  # 说明边框参数作用
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))  # 浅黄色背景框

# ========== 子图 J：横向散点条图 ==========
ax = axes1[1, 4]  # 获取第2行第5列子图对象
sns.stripplot(data=tips,  # 数据源
              x='total_bill',  # x轴：账单总额（数值）
              y='day',  # y轴：星期（分类）-交换实现横向
              hue='day',  # 分组：按星期
              palette='rocket',  # rocket红黑渐变调色板
              legend=False,  # 隐藏图例
              jitter=0.2,  # jitter=0.2：垂直抖动范围0.2
              size=5,  # 点大小：5点
              ax=ax)  # 绘制目标
ax.set_title('J. 横向散点条图\n交换x和y', fontsize=11, fontweight='bold')  # 标题说明横向实现
ax.set_xlabel('账单金额 ($)', fontsize=10)  # x轴标签
ax.set_ylabel('星期', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        '参数: x和y互换\njitter垂直抖动\n适合类别名长',  # 说明横向参数和适用场景
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))  # 浅蓝色背景框

plt.tight_layout()  # 自动调整子图间距，避免重叠
plt.show()  # 显示画布1的10个子图
```

```python
# ============================================================
# 画布 2: 组合技巧和实用案例
# ============================================================
fig2, axes2 = plt.subplots(2, 5, figsize=(24, 10))  # 创建2行5列子图布局，总画布大小24x10英寸
fig2.suptitle('Seaborn Stripplot 教程 - 画布2: 组合技巧与实用案例',  # 设置整个画布的总标题
              fontsize=16,  # 总标题字体大小16
              fontweight='bold',  # 总标题加粗显示
              y=0.995)  # 总标题垂直位置略低于顶部（避免裁剪）

# ========== 子图 K：Boxplot + Stripplot ==========
ax = axes2[0, 0]  # 获取第1行第1列子图对象
sns.boxplot(data=penguins,  # 数据源：企鹅数据集
            x='species',  # x轴：物种
            y='body_mass_g',  # y轴：体重
            hue='species',  # 分组：按物种着色
            palette='pastel',  # 柔和色彩调色板
            legend=False,  # 隐藏图例
            width=0.5,  # 箱体宽度：0.5
            ax=ax)  # 绘制箱线图
sns.stripplot(data=penguins,  # 数据源
              x='species',  # x轴
              y='body_mass_g',  # y轴
              color='black',  # 散点颜色：黑色
              alpha=0.3,  # 散点透明度：30%
              size=3,  # 散点大小：3点
              ax=ax)  # 叠加到同一子图
ax.set_title('K. Box + Strip组合\n统计+原始数据', fontsize=11, fontweight='bold')  # 标题说明经典组合
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        '经典组合:\nboxplot概览\nstripplot细节\n最常用搭配',  # 说明最常用组合优势
        transform=ax.transAxes,
        fontsize=7,  # 字体小（内容多）
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))  # 浅黄色背景框

# ========== 子图 L：Violin + Stripplot ==========
ax = axes2[0, 1]  # 获取第1行第2列子图对象
sns.violinplot(data=tips,  # 数据源：tips数据集
               x='day',  # x轴：星期
               y='tip',  # y轴：小费
               hue='day',  # 分组：按星期
               palette='mako',  # mako蓝绿渐变调色板
               legend=False,  # 隐藏图例
               inner=None,  # 无内部标记
               alpha=0.6,  # 小提琴透明度：60%
               ax=ax)  # 绘制小提琴图
sns.stripplot(data=tips,  # 数据源
              x='day',  # x轴
              y='tip',  # y轴
              color='red',  # 散点颜色：红色
              alpha=0.4,  # 散点透明度：40%
              size=3,  # 散点大小：3点
              ax=ax)  # 叠加散点图
ax.set_title('L. Violin + Strip组合\n密度+数据点', fontsize=11, fontweight='bold')  # 标题说明密度组合
ax.set_xlabel('星期', fontsize=10)  # x轴标签
ax.set_ylabel('小费 ($)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        '组合技巧:\nviolin分布\nstrip原始点\n双重验证',  # 说明双重验证优势
        transform=ax.transAxes,
        fontsize=7,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))  # 浅绿色背景框

# ========== 子图 M：Barplot + Stripplot ==========
ax = axes2[0, 2]  # 获取第1行第3列子图对象
sns.barplot(data=penguins,  # 数据源
            x='island',  # x轴：岛屿
            y='flipper_length_mm',  # y轴：鳍长
            hue='island',  # 分组：按岛屿
            palette='Set3',  # Set3分类调色板
            legend=False,  # 隐藏图例
            alpha=0.7,  # 柱状图透明度：70%
            errorbar='sd',  # 误差条：标准差
            ax=ax)  # 绘制柱状图
sns.stripplot(data=penguins,  # 数据源
              x='island',  # x轴
              y='flipper_length_mm',  # y轴
              color='darkblue',  # 散点颜色：深蓝色
              alpha=0.3,  # 散点透明度：30%
              size=4,  # 散点大小：4点
              jitter=0.2,  # 水平抖动：0.2
              ax=ax)  # 叠加散点图
ax.set_title('M. Bar + Strip组合\n均值+分散', fontsize=11, fontweight='bold')  # 标题说明均值分散组合
ax.set_xlabel('岛屿', fontsize=10)  # x轴标签
ax.set_ylabel('鳍长 (mm)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        '组合用途:\nbar显示均值\nstrip显示分散\n避免误导',  # 说明避免误导优势
        transform=ax.transAxes,
        fontsize=7,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightcyan', alpha=0.5))  # 浅青色背景框

# ========== 子图 N：按数值排序 ==========
ax = axes2[0, 3]  # 获取第1行第4列子图对象
species_order = penguins.groupby('species')['bill_depth_mm'].median().sort_values().index  # 按物种分组，计算喙深中位数，升序排序，返回索引
sns.stripplot(data=penguins,  # 数据源
              x='species',  # x轴：物种
              y='bill_depth_mm',  # y轴：喙深
              order=species_order,  # order参数：按中位数排序
              hue='species',  # 分组
              palette='viridis',  # viridis连续色彩调色板
              legend=False,  # 隐藏图例
              size=5,  # 散点大小：5点
              alpha=0.6,  # 透明度：60%
              ax=ax)  # 绘制目标
ax.set_title('N. 按中位数排序\norder参数', fontsize=11, fontweight='bold')  # 标题说明排序依据
ax.set_xlabel('企鹅种类(按中位数排序)', fontsize=10)  # x轴标签说明排序方法
ax.set_ylabel('喙深 (mm)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'order用法:\n按median排序\n便于观察趋势',  # 说明排序趋势观察优势
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lavender', alpha=0.5))  # 淡紫色背景框

# ========== 子图 O：自定义颜色字典 ==========
ax = axes2[0, 4]  # 获取第1行第5列子图对象
custom_colors = {  # 自定义颜色字典：花种→十六进制颜色
    'setosa': '#FF6B6B',  # 山茶花：珊瑚红
    'versicolor': '#4ECDC4',  # 变色鸢尾：青绿色
    'virginica': '#95E1D3'  # 弗吉尼亚鸢尾：浅青色
}
sns.stripplot(data=iris,  # 数据源：iris数据集
              x='species',  # x轴：花种
              y='petal_length',  # y轴：花瓣长度
              hue='species',  # 分组
              palette=custom_colors,  # 使用自定义颜色字典
              legend=False,  # 隐藏图例
              size=6,  # 散点大小：6点
              alpha=0.7,  # 透明度：70%
              ax=ax)  # 绘制目标
ax.set_title('O. 自定义颜色映射\npalette dict (Iris)', fontsize=11, fontweight='bold')  # 标题说明字典配色
ax.set_xlabel('鸢尾花种类', fontsize=10)  # x轴标签
ax.set_ylabel('花瓣长度 (cm)', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        'palette字典:\n精确控制颜色\n品牌配色',  # 说明字典精确控制优势
        transform=ax.transAxes,
        fontsize=8,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='mistyrose', alpha=0.5))  # 玫瑰色背景框

# ========== 子图 P：分组dodge + 不同marker ==========
ax = axes2[1, 0]  # 获取第2行第1列子图对象
# 分别绘制两个性别，使用不同marker
for sex, marker in zip(['Male', 'Female'], ['o', '^']):  # 循环：性别对应圆形/三角形
    data_subset = penguins[penguins['sex'] == sex]  # 筛选当前性别数据
    sns.stripplot(data=data_subset,  # 子集数据
                  x='species',  # x轴：物种
                  y='body_mass_g',  # y轴：体重
                  marker=marker,  # 当前性别标记形状
                  label=sex,  # 图例标签
                  size=6,  # 散点大小：6点
                  alpha=0.6,  # 透明度：60%
                  ax=ax)  # 叠加到同一子图
ax.set_title('P. 不同组不同形状\n多次调用stripplot', fontsize=11, fontweight='bold')  # 标题说明多维技巧
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.set_ylabel('体重 (g)', fontsize=10)  # y轴标签
ax.legend(title='性别', loc='upper left')  # 图例设置
ax.text(0.02,  # 文本位置
        0.98,
        '技巧:\n循环调用\n每组不同marker\n更多维度信息',  # 说明多维度编码技巧
        transform=ax.transAxes,
        fontsize=7,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='peachpuff', alpha=0.5))  # 桃色背景框

# ========== 子图 Q：大数据集降采样 ==========
ax = axes2[1, 1]  # 获取第2行第2列子图对象
# 创建大数据集
np.random.seed(42)  # 设置随机种子确保可重复
large_data = pd.DataFrame({  # 创建1000行大数据
    'category': np.random.choice(['A', 'B', 'C'], 1000),  # 随机类别
    'value': np.random.randn(1000) * 10 + 50  # 正态分布数值
})
# 降采样
sampled_data = large_data.groupby('category').sample(n=50, random_state=42)  # 每组随机抽取50个样本
sns.stripplot(data=sampled_data,  # 降采样数据
              x='category',  # x轴：类别
              y='value',  # y轴：数值
              hue='category',  # 分组
              palette='deep',  # deep深色调色板
              legend=False,  # 隐藏图例
              alpha=0.5,  # 透明度：50%
              size=4,  # 散点大小：4点
              ax=ax)  # 绘制目标
ax.set_title('Q. 大数据集降采样\nsample(n=50)', fontsize=11, fontweight='bold')  # 标题说明降采样
ax.set_xlabel('类别', fontsize=10)  # x轴标签
ax.set_ylabel('数值', fontsize=10)  # y轴标签
ax.text(0.02,  # 文本位置
        0.98,
        '优化技巧:\n数据量大时\n按组降采样\n保持代表性',  # 说明降采样保持代表性
        transform=ax.transAxes,
        fontsize=7,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))  # 浅黄色背景框

# ========== 子图 R：双Y轴对比 ==========
ax = axes2[1, 2]  # 获取第2行第3列子图对象
ax_twin = ax.twinx()  # 创建右Y轴
# 左Y轴
sns.stripplot(data=penguins,  # 数据源
              x='species',  # x轴：物种
              y='body_mass_g',  # 左Y轴：体重
              color='blue',  # 蓝色散点
              alpha=0.5,  # 透明度：50%
              size=4,  # 散点大小：4点
              ax=ax)  # 绘制到左轴
ax.set_ylabel('体重 (g)', fontsize=10, color='blue')  # 左Y轴标签
ax.tick_params(axis='y', labelcolor='blue')  # 左Y轴刻度颜色
# 右Y轴
sns.stripplot(data=penguins,  # 数据源
              x='species',  # x轴
              y='flipper_length_mm',  # 右Y轴：鳍长
              color='red',  # 红色散点
              alpha=0.5,  # 透明度：50%
              size=4,  # 散点大小：4点
              marker='^',  # 三角形标记
              ax=ax_twin)  # 绘制到右轴
ax_twin.set_ylabel('鳍长 (mm)', fontsize=10, color='red')  # 右Y轴标签
ax_twin.tick_params(axis='y', labelcolor='red')  # 右Y轴刻度颜色
ax.set_title('R. 双Y轴对比\n两个变量', fontsize=11, fontweight='bold')  # 标题说明双变量对比
ax.set_xlabel('企鹅种类', fontsize=10)  # x轴标签
ax.text(0.02,  # 文本位置
        0.98,
        '高级技巧:\ntwinx()双Y轴\n对比两个变量\n不同marker区分',  # 说明双Y轴技巧
        transform=ax.transAxes,
        fontsize=7,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))  # 浅蓝色背景框

# ========== 子图 S：时间序列散点 ==========
ax = axes2[1, 3]  # 获取第2行第4列子图对象
# 创建时间序列数据
dates = pd.date_range('2024-01-01', periods=100, freq='D')  # 2024年1月1日起100天日期
ts_data = pd.DataFrame({  # 创建时间序列DataFrame
    'date': dates,  # 日期列
    'value': np.cumsum(np.random.randn(100)) + 100,  # 累积和+随机游走
    'category': np.random.choice(['High', 'Low'], 100)  # 随机高低类别
})
# 只取部分点避免过密
ts_sample = ts_data.iloc[::5]  # 每5个取1个（降采样20个点）
sns.stripplot(data=ts_sample,  # 降采样数据
              x='date',  # x轴：日期
              y='value',  # y轴：数值
              hue='category',  # 分组：高低类别
              palette='Set1',  # Set1分类调色板
              size=6,  # 散点大小：6点
              alpha=0.7,  # 透明度：70%
              ax=ax)  # 绘制目标
ax.set_title('S. 时间序列散点\n降采样展示', fontsize=11, fontweight='bold')  # 标题说明时间序列
ax.set_xlabel('日期', fontsize=10)  # x轴标签
ax.set_ylabel('数值', fontsize=10)  # y轴标签
plt.setp(ax.xaxis.get_majorticklabels(), rotation=45)  # x轴标签旋转45度
ax.legend(title='类别', loc='upper left', fontsize=8)  # 图例设置
ax.text(0.02,  # 文本位置
        0.98,
        '应用场景:\n时间趋势\n降采样避免拥挤\n保留异常值',  # 说明时间序列应用
        transform=ax.transAxes,
        fontsize=7,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))  # 浅黄色背景框

# ========== 子图 T：综合美化案例 ==========
ax = axes2[1, 4]  # 获取第2行第5列子图对象
sns.violinplot(data=tips,  # 数据源
               x='day',  # x轴：星期
               y='total_bill',  # y轴：账单总额
               hue='time',  # 分组：时间段
               split=True,  # 左右分割显示
               palette='Set2',  # Set2柔和色彩调色板
               inner='quartile',  # 显示四分位线
               linewidth=1.5,  # 边框线宽：1.5点
               ax=ax)  # 绘制小提琴图
# 叠加stripplot，使用dodge匹配violin的分组
sns.stripplot(data=tips,  # 数据源
              x='day',  # x轴
              y='total_bill',  # y轴
              hue='time',  # 分组：时间段
              dodge=True,  # dodge=True：与violin分组对齐
              palette='dark:black',  # 深色黑色调色板
              alpha=0.3,  # 透明度：30%
              size=2,  # 小点大小：2点
              ax=ax,  # 叠加到同一子图
              legend=False)  # 隐藏重复图例
ax.set_title('T. 综合美化案例\nViolin+Strip精细组合', fontsize=11, fontweight='bold')  # 标题说明精细组合
ax.set_xlabel('星期', fontsize=10)  # x轴标签
ax.set_ylabel('账单金额 ($)', fontsize=10)  # y轴标签
ax.legend(title='时间段',  # 图例标题
          frameon=True,  # 显示边框
          fancybox=True,  # 圆角边框
          shadow=True)  # 阴影效果
ax.grid(axis='y', alpha=0.3, linestyle='--')  # y轴虚线网格，透明度30%
ax.text(0.02,  # 文本位置（垂直居中）
        0.50,
        '综合技巧:\n split violin\ndodge strip\n 半透明点\n 小点大小\n 美化图例',  # 总结5大美化技巧
        transform=ax.transAxes,
        fontsize=7,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))  # 浅绿色背景框

plt.tight_layout()  # 自动调整子图间距，避免重叠
plt.show()  # 显示画布2的10个子图
```

### 蜂群图-swarmplot()函数
swarmplot()函数用于创建蜂群图（swarm plot），显示分类变量与数值变量之间的关系。它通过调整数据点的位置来避免重叠，从而更清晰地展示数据的分布情况。swarmplot()函数适用于较小的数据集，因为它会根据数据点的数量自动调整位置，以确保每个数据点都能被清晰地看到。
#### swarmplot()函数参数
- x, y, hue：用于指定分类变量和数值变量。x和y，分别表示分类变量和数值变量，hue用于指定分组变量。
- order, hue_order：用于指定分类变量和分组变量的绘制顺序
- palette：用于指定颜色调色板。
- size：用于指定数据点的大小。
- marker：用于指定数据点的形状, 参数有'D'为菱形，'o'为圆, 's', '^'。
- alpha：用于指定数据点的透明度。
- edgecolor：用于指定数据点的边缘颜色。
- linewidth：用于指定数据点的边缘宽度。
- hue：用于指定分组变量。
- warn_thresh:用于指定警告阈值。当数据点的数量超过该阈值时，swarmplot函数会发出警告，提示用户数据点可能会重叠。默认情况下，warn_thresh=512。
- ax：用于指定绘图的坐标轴对象。
- dodge：用于指定是否对分组数据进行错开显示, 参数为True或False为分开和不分开。
- legend：用于指定是否显示图例。

### catplot()函数
catplot()返回的是一个FacetGrid对象，方便进行多图布局调整。<font color='red'>没有办法直接在已有的ax上绘图</font>。
catplot()函数是Seaborn库中的一个高级绘图函数，用于创建分类图。它可以绘制多种类型的分类图，如条形图（bar plot）、箱线图（box plot）、小提琴图（violin plot）等，取决于传递的参数。catplot()函数支持多维数据映射，通过hue等参数，可以在同一图形中展示多个变量之间的关系。此外，catplot()还支持创建图形网格，允许在多个子图中显示数据，从而更好地比较不同类别或条件下的数据分布和趋势。
#### sns.catplot(data, x, y, kind='bar')相关参数
- x: x轴数据
- y: y轴数据
- kind: 图表类型，可以是'bar'（条形图）、'box'（箱线图）、'violin'（小提琴图）、'strip'（抖动散点图）、'swarm'（蜂群图）等，按照其形状对应相应的参数参考barplot()、boxplot()、violinplot()、stripplot()或swarmplot()函数的参数
- hue: 用于区分数据点颜色的变量
- col: 用于创建列子图的变量,分为多列显示
- row: 用于创建行子图的变量,分为多行显示
- col_wrap: 用于设置列子图的换行数量,假如col有12列且col_wrap=4,则会显示3行4列的子图
- row_order: 用于设置行子图的顺序
- col_order: 用于设置列子图的顺序
- margin_titles: 是否在子图上方显示标题，默认为False

### 热力图-heatmap()函数和clustermap()函数
heatmap()函数用于创建热力图，显示二维数据的数值大小。它通过颜色的深浅来表示数值的大小，适用于展示矩阵数据或相关性矩阵等。
clustermap()函数用于创建聚类热力图，结合了热力图和层次聚类分析。它通过对数据进行聚类分析，将相似的数据点分组，并在热力图中显示这些分组的关系。
#### heatmap()函数参数
- data：用于指定绘图的数据。它可以是一个二维数组、DataFrame或类似的结构。
- annot：用于指定是否在热力图上显示数值注释。默认情况下，annot=False，表示不显示注释。可以将其设置为True，以显示注释。
- annot_kws：用于指定注释的样式属性。它是一个字典，可以包含字体大小、颜色等属性。
- fmt：用于指定数值注释的格式。默认情况下，fmt='.2g'，表示使用通用格式显示数值。可以将其设置为其他格式字符串，以自定义数值的显示方式。
- cmap：用于指定颜色映射表。它可以是预定义的颜色映射名称（如'viridis'、'plasma'等）或自定义的颜色列表。可选的值包括'magma'、'cividis'、'coolwarm'、'YlGnBu'等。分别为不同的颜色渐变效果。
- cbar：用于指定是否显示颜色条。默认情况下，cbar=True，表示显示颜色条。可以将其设置为False，以隐藏颜色条。
- cbar_kws：用于指定颜色条的样式属性。它是一个字典，可以包含标签、刻度等属性。
- cbar_ax：用于指定颜色条的坐标轴对象。默认情况下，cbar_ax=None，表示在当前坐标轴上绘制颜色条。可以将其设置为一个matplotlib的Axes对象，以在指定的坐标轴上绘制颜色条.
- vmin, vmax：用于指定颜色映射的最小值和最大值。
- linewidths：用于指定单元格之间的线宽。默认情况下，linewidths=0，表示不显示线条。可以将其设置为正数，以显示单元格之间的线条。
- linecolor：用于指定单元格之间线条的颜色。默认情况下，linecolor='white'，表示线条为白色。可以将其设置为其他颜色名称或十六进制颜色代码。
- square：用于指定单元格是否为正方形。默认情况下，square=False，表示单元格不是正方形。可以将其设置为True，以使单元格为正方形。
- robust：用于指定是否使用稳健的颜色映射。默认情况下，robust=False，表示不使用稳健的颜色映射。可以将其设置为True，以使用稳健的颜色映射，从而减少异常值对颜色映射的影响。
- mask：用于指定要遮罩的单元格。它是一个与数据形状相同的布尔数组，其中True表示要遮罩的单元格，False表示不遮罩的单元格。
- xticklabels, yticklabels：用于指定x轴和y轴的刻度标签。它们可以是布尔值、列表或整数。默认情况下，xticklabels和yticklabels=True，表示显示所有刻度标签。可以将它们设置为False，以隐藏刻度标签；也可以将它们设置为一个列表，以自定义刻度标签；还可以将它们设置为一个整数，以指定要显示的刻度标签的数量。

#### melt()函数和pivot()函数
这两个函数是数据处理和变换中非常重要的工具。是相反的操作。  
melt()函数用于将数据从宽格式转换为长格式。它通过将多个列合并为两个列（变量名和变量值）来实现这一转换，适用于需要进行分组或聚合分析的数据处理。   
pivot()函数用于将数据从长格式转换为宽格式。它通过将一个列的唯一值转换为多个列来实现这一转换，适用于需要重新组织数据以便进行分析或可视化的场景。

#### np.triu(m, k)相关参数
上三角为True下面为False
- m：输入的二维数组（矩阵）。
- k：对角线的偏移量。k=0表示主对角线，k>0表示上方第k条对角线，k<0表示下方第k条对角线。k=1表示主对角线以上一条对角线，k=-1表示主对角线以下一条对角线, k=0时再对角线，每+1向上平移，-1向下平移。  
#### np.tril(m, k)相关参数
下三角为True上面为False
- m：输入的二维数组（矩阵）。
- k：对角线的偏移量。k=0表示主对角线，k>0表示上方第k条对角线，k<0表示下方第k条对角线。  
#### 与mask参数的联系
seaborn.heatmap 中 mask 参数的<font color='red'>设计逻辑是相反</font>的False将会被显示：
mask 中为 True 的单元格，数据将被“遮盖”（隐藏），不会被绘制出来。
mask 中为 False 的单元格，数据才会被“显示”（绘制）。

```python
import numpy as np
#上三角为True
matrix = np.triu(np.ones((8, 8)), k=0).astype(bool)
print(matrix)
#下三角为True
print()
matrix = np.tril(np.ones((8, 8)), k=0).astype(bool)
print(matrix)
```

#### 混淆矩阵
混淆矩阵（Confusion Matrix）是一种用于评估分类模型性能的工具。它以矩阵的形式展示了模型在不同类别上的预测结果与实际结果之间的关系。混淆矩阵通常用于二分类和多分类问题，通过比较模型的预测结果与真实标签，帮助我们了解模型的准确性、精确率、召回率等指标。  
**混淆矩阵的核心**：四个指标，T：True（真实），F：False（错误），P：Positive（正类），N：Negative（负类）  
- 真阳性（True Positive, TP）：模型正确地将正类样本预测为正类的数量。
- 假阳性（False Positive, FP）：模型错误地将负类样本预测为正类的数量。
- 真阴性（True Negative, TN）：模型正确地将负类样本预测为负类的数量。
- 假阴性（False Negative, FN）：模型错误地将正类样本预测为负类的数量。

**用处**：计算多种关键性能指标  
1. 准确率（Accuracy）：衡量模型整体预测的正确性，计算公式为 (TP + TN) / (TP + TN + FP + FN)。
2. 精确率（Precision）：衡量模型在预测为正类时的准确性，计算公式为 TP / (TP + FP)。
3. 召回率（Recall）：衡量模型对正类样本的识别能力，计算公式为 TP / (TP + FN)。
4. F1分数（F1 Score）：精确率和召回率的调和平均数，计算公式为 2 * (Precision * Recall) / (Precision + Recall)。

## 分布图
分布图是一种用于显示数据分布情况的图表。它通过展示数据的频率或概率密度来揭示数据的分布特征。分布图常用于分析数据的集中趋势、离散程度和形态，帮助识别数据中的模式和异常值。

### 直方图-histplot()函数
histplot()函数用于创建直方图，显示数据的分布情况。它通过将数据划分为多个区间（bins）并计算每个区间内的数据点数量来绘制直方图，适用于连续变量的数据可视化。
#### histplot()函数参数
- x, y：用于指定绘图的数据。x表示要绘制直方图的变量，y表示要绘制的频数或密度。
- bins：用于指定直方图的区间数量或区间边界。它可以是一个整数，表示区间的数量；也可以是一个序列，表示区间的边界值。
- binwidth：用于指定每个区间的宽度。它是一个正数，表示每个区间的宽度。
- binrange=[min, max]：用于指定直方图的范围。它是一个元组，包含两个值，分别表示直方图的最小值和最大值。
- cumulative：用于指定是否绘制累积直方图。默认情况下，cumulative=False，表示绘制普通直方图。可以将其设置为True，以绘制累积直方图。
- stat：用于指定直方图的统计量类型。可选的值包括'count'（频数）、'frequency'（频率）、'density'（密度）和'probability'（概率）。
- common_bins：用于指定是否在多个子图中使用相同的区间边界。默认情况下，common_bins=False，表示每个子图使用独立的区间边界。可以将其设置为True，以在所有子图中使用相同的区间边界。
- multiple：用于指定直方图的绘制方式。可选的值包括'layer'（层叠）、'dodge'（错开）和'stack'（堆叠）, 'fill'（填充）。
- kde: 用于指定是否在直方图上绘制核密度估计曲线。默认情况下，kde=False，表示不绘制核密度估计曲线。可以将其设置为True，以绘制核密度估计曲线。
- bw_adjust：用于调整核密度估计的带宽。它是一个乘数，用于调整默认带宽的大小。较大的bw_adjust值会增加带宽，从而使密度估计更平滑；较小的bw_adjust值会减小带宽，从而使密度估计更尖锐。
- element：用于指定直方图的元素类型。可选的值包括'bars'（条形）、'step'（阶梯）和'poly'（多边形）。
- fill：用于指定是否填充直方图的条形。默认情况下，fill=True，表示填充条形。可以将其设置为False，以仅绘制条形的边框。
- shrink：用于指定条形的缩放比例。它是一个介于0和1之间的浮点数，表示条形的宽度相对于区间宽度的比例。
- discrete：用于指定是否将数据视为离散变量。默认情况下，discrete=False，表示数据为连续变量。可以将其设置为True，以将数据视为离散变量。
- thresh: 用于指定显示条形的最小高度阈值。默认情况下，thresh=0，表示显示所有条形。可以将其设置为正数，以仅显示高度大于该阈值的条形。过滤掉噪声。

####  cmap和 palette区别
cmap（colormap）和palette都是用于指定颜色的工具，但它们在使用场景和功能上有所不同。
- cmap='mako'：用于连续的数值数据（例如：温度、高度、数量）。
- palette='Set2'：用于分类的数据（例如：性别、物种、国家）。
- cmap：cmap通常用于<font color='red'>连续数据</font>的颜色映射。它是一种颜色渐变表，可以根据数据的数值范围将颜色映射到不同的数值上。cmap通常用于热力图、散点图等需要表示连续数值的数据可视化中。常见的cmap有'viridis'、'plasma'、'inferno'等。
- palette：palette通常用于<font color='red'>分类数据</font>的颜色选择。它是一组预定义的颜色集合，可以为不同的类别分配不同的颜色。palette通常用于条形图、箱线图等需要区分不同类别的数据可视化中。常见的palette有'deep'、'muted'、'bright'等。

### ecdfplot()函数
ecdfplot()函数用于创建经验累积分布函数（Empirical Cumulative Distribution Function, ECDF）图，显示数据的累积分布情况。它通过计算每个数据点的累积频率来绘制ECDF图，适用于连续变量的数据可视化。
#### ecdfplot()函数参数
- x, y：用于指定绘图的数据。x表示要绘制ECDF的变量，y表示要绘制的频数或密度。
- complementary：用于指定是否绘制补充ECDF。默认情况下，complementary=False，表示绘制普通ECDF。可以将其设置为True，以绘制补充ECDF。
- stat：用于指定ECDF的统计量类型。可选的值包括'proportion'（比例）、'count'（频数）和'density'（密度）。
- log_scale：用于指定是否对轴进行对数缩放。它可以是一个布尔值或包含轴名称的列表。
- weights：用于指定数据点的权重。它是一个与数据长度相同的数组，表示每个数据点的权重。
- label：用于指定图例标签。它是一个字符串，表示图例中显示的标签名称。
#### vlines()和hlines()函数
vlines()函数用于在图形中绘制垂直线，hlines()函数用于绘制水平线。它们可以用于突出显示特定的数值或位置
- vlines(x, ymin, ymax, colors, linestyles, label)：用于绘制垂直线。参数包括x（线的位置）、ymin和ymax（线的起始和结束位置）、colors（线的颜色）、linestyles（线的样式）和label（线的标签）。
- hlines(y, xmin, xmax, colors, linestyles, label)：用于绘制水平线。参数包括y（线的位置）、xmin和xmax（线的起始和结束位置）、colors（线的颜色）、linestyles（线的样式）和label（线的标签）。

```python
#ECDF原理和计算
fig3, ax3 = plt.subplots(1, 2, figsize=(16, 6))
fig3.suptitle(
    "ECDF的原理和计算",
    fontsize=14,
    fontweight='bold'
)
#左图：小样本展示ECDF的阶梯特性
ax = ax3[0]
small_sample = penguins.sample(n=20, random_state=42)['body_mass_g'].sort_values()#随机抽样20个体重数据并排序
print(small_sample.shape)
for i, value in enumerate(small_sample):
    proportion = (i+1)/len(small_sample)
    if i > 0:
        ax.hlines( #水平线
            proportion,
            small_sample.iloc[i-1],
            value, #当前值
            colors='blue',
            linewidth=2
        )
        ax.vlines( #垂直线
            value,
            proportion - 1/len(small_sample),
            proportion,
            colors='blue',
            linewidth=2,
            linestyle='dashed', #虚线
            alpha=0.5
        )
        ax.plot(
            value,
            proportion,
            'ro', #红色点
            markersize=5
        )
ax.set_title(
    'ECDF阶段函数特性\n小样本(n=20)',
    fontsize=12,
    fontweight='bold'
)
ax.set_xlabel('体重 (g)', fontsize=10)
ax.set_ylabel('累积概率', fontsize=10)
ax.grid(alpha=0.3)
ax.text(
    0.02,
    0.98,
    'ECDF特点：\n阶梯函数\n显示数据分布\n累积概率',  # 说明ECDF作用
    transform=ax.transAxes,
    fontsize=8,
    verticalalignment='top',
    bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5)
)
#右图：大样本展示ECDF的平滑特性
ax = ax3[1]
for n in [20, 50, 100, 333]:
    sample = penguins.sample(n=n, random_state=42)
    sns.ecdfplot(
        data=sample,
        x='body_mass_g',
        ax=ax,
        label=f'n={n}',
        alpha=0.7
    )
ax.set_title('样本量对ECDF平滑性的影响\n大样本', fontsize=12, fontweight='bold')
ax.set_xlabel('体重 (g)', fontsize=10)
ax.set_ylabel('累积概率', fontsize=10)
ax.grid(alpha=0.3)
ax.legend(title='样本量', fontsize=8)
ax.text(
    0.02,
    0.98,
    '样本量影响：\n样本量越大\nECDF越平滑\n更好反映分布',  # 说明ECDF
    transform=ax.transAxes,
    fontsize=8,
    verticalalignment='top',
    bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5)
)
plt.tight_layout()
plt.show()
```

### Axes-level functions (坐标轴级别函数)和 Figure-level functions (图形级别函数)
😶‍🌫️Axes-level functions (坐标轴级别函数) 是指那些直接在单个坐标轴（Axes）上绘制图形的函数。这些函数通常用于创建简单的图形，如散点图、折线图、柱状图等。它们允许用户对单个图形进行详细的控制和定制。常见的Axes-level functions包括 scatterplot()、lineplot()、barplot()等。  
😶‍🌫️Figure-level functions (图形级别函数) 是指那些用于创建整个图形（Figure）的函数。这些函数通常用于创建复杂的图形布局，如多子图、分面图等。它们允许用户在一个图形中包含多个坐标轴，并对整个图形进行统一的控制和定制。常见的Figure-level functions包括 relplot()、catplot()、pairplot()等。
#### 如何选择使用哪种函数
- 如果需要创建简单的图形，并且只涉及单个坐标轴，建议使用Axes-level functions。这些函数提供了更细粒度的控制，适合对图形进行详细的定制。
- 如果需要创建复杂的图形布局，或者需要在一个图形中包含多个坐标轴，建议使用Figure-level functions。这些函数提供了更高层次的抽象，适合创建复杂的图形结构。
#### 判断
如果一个 Seaborn 函数能用 col 或 row 参数来自动创建网格，或者它本身就是由多个子图（如 jointplot, pairplot）构成的，那么它返回的就是一个 Grid 对象，而不是 ax。
#### 方法
##### sns.pairplot()函数返回PairGrid对象后
- 使用 `map_diag()` 方法可以在对角线上绘制图形。
- 使用 `map_offdiag()` 方法可以在非对角线上绘制图形。
- 使用 `map_lower()` 方法可以设置子图的标题。
- 使用 `map_upper()` 方法可以设置子图的标题。
##### sns.jointplot()	JointGrid对象后
- 使用 `plot_joint()` 方法可以在主图区域绘制图形。
- 使用 `plot_marginals()` 方法可以在边缘区域绘制图形。
##### sns.relplot()函数返回FacetGrid对象后
- 使用 `map()` 方法可以在整个网格上绘制图形。
- 使用 `map_dataframe()` 方法可以在整个网格上绘制数据框中的列。

### jointplot
jointplot()函数用于创建联合分布图，显示两个变量之间的关系以及它们各自的分布情况。它通过绘制散点图和边际分布图来展示变量之间的关系和分布特征，适用于探索两个连续变量之间的关系。其返回的是一个JointGrid对象，允许进一步自定义图形。
#### jointplot()函数参数
- x, y：用于指定绘图的数据。x表示要绘制的第一个变量，y表示要绘制的第二个变量。
- kind：用于指定联合分布图的类型。可选的值包括'scatter'（散点图）、'kde'（核密度估计图）、'hist'（直方图）和'reg'（回归图）。
- height：用于指定图形的高度。它是一个正数，表示图形的高度，单位为英寸。
- ratio：用于指定边际分布图与联合分布图的比例。默认情况下，ratio=5，表示边际分布图的高度为联合分布图高度的1/5。可以将其设置为其他正数，以调整边际分布图的高度比例。
- space：用于指定边际分布图与联合分布图之间的间距。它是一个正数，表示间距的大小，单位为英寸。
- kind: 用于指定联合分布图的类型。可选的值包括'scatter'（散点图）、'kde'（核密度估计图）、'hist'（直方图）和'reg'（回归图）。
- bins：用于指定边际分布图的区间数量或区间边界。它可以是一个整数，表示区间的数量；也可以是一个序列，表示区间的边界值。
- marginal_kws：用于指定边际分布图的样式属性。它是一个字典，可以包含颜色、线条样式等属性。


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
sns.set_style('whitegrid')
plt.rcParams['font.family'] = ['SimHei','Times New Roman']  #设置
penguins = sns.load_dataset('penguins').dropna()  # 加载企鹅数据集并删除缺失值
g = sns.jointplot(
    data=penguins,
    x='flipper_length_mm',
    y='body_mass_g',
    kind='hex', #六边形箱
    # cmap='viridis',
    height=8,
)
g.set_axis_labels('鳍长 (mm)', '体重 (g)', fontsize=12) #设置轴标签
g.fig.suptitle('Seaborn Jointplot - 六边形箱示例', fontsize=16, fontweight='bold', y=1.02)
```

```python
g = sns.jointplot(
    data=penguins,
    x='flipper_length_mm',
    y='body_mass_g',
    kind='scatter', #散点图
    # cmap='viridis',
    hue='species',
    palette='Set2',
    height=8,
)
g.set_axis_labels('鳍长 (mm)', '体重 (g)', fontsize=12) #设置轴标签
g.fig.suptitle('Seaborn Jointplot - 六边形箱示例', fontsize=16, fontweight='bold', y=1.02)
```

### pariplot()函数
pariplot()函数用于创建成对关系图，显示多个变量之间的两两关系。它通过绘制散点图矩阵来展示变量之间的关系，适用于探索多个连续变量之间的相互关系。
#### pairplot()函数参数
- data：用于指定绘图的数据。它可以是一个DataFrame或类似的结构。
- hue：用于指定分组变量。可以将数据按照hue变量的不同取值进行分组，并为每个组绘制不同的颜色。默认情况下，hue=None，表示不进行分组。
- palette：用于指定颜色调色板。可以设置为预定义的调色板名称（如'deep'、'muted'、'bright'等），也可以设置为自定义的颜色列表。默认情况下，palette=None，表示使用默认的颜色调色板。
- diag_kind：用于指定对角线图的类型。可选的值包括'hist'（直方图）和'kde'（核密度估计图）。默认情况下，diag_kind='auto'，表示根据数据类型自动选择对角线图的类型。
- markers：用于指定散点图的标记样式。它可以是一个字符串或一个列表，表示散点图中数据点的标记样式。默认情况下，markers=None，表示使用默认的标记样式。
- vars：用于指定要绘制的变量列表。默认情况下，vars=None，表示绘制所有变量。可以将其设置为一个变量列表，以仅绘制指定的变量。
- plot_kws：用于指定散点图的样式属性。它是一个字典，可以包含颜色、大小、透明度等属性。
- diag_kws：用于指定对角线图的样式属性。它是一个字典，可以包含颜色、线条样式等属性,参数可以为直方图或核密度估计图的样式。hist为直方图，kde为核密度估计图。
- corner=True: 用于指定是否仅绘制矩阵的下三角部分。默认情况下，corner=False，表示绘制整个矩阵。可以将其设置为True，以仅绘制下三角部分，从而减少重复信息的显示。

```python
import seaborn as sns 
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
sns.set_style('whitegrid')
plt.rcParams['font.family'] = ['SimHei', 'Times New Roman']
np.random.seed(42)

penguins = sns.load_dataset('penguins').dropna() #加载企鹅数据级
tips = sns.load_dataset('tips').dropna() #加载小费数据集
diamonds = sns.load_dataset('diamonds').sample(n=500, random_state=42) #加载钻石数据集的子集
print(diamonds.columns.tolist())
print(tips.columns.tolist())
print(penguins.columns.tolist())
students = pd.DataFrame({
    '语文':np.random.randint(60, 100, 100),
    '数学':np.random.randint(60, 100, 100),
    '英语':np.random.randint(60, 100, 100),
    '物理':np.random.randint(60, 100, 100),
    '化学':np.random.randint(60, 100, 100),
    '生物':np.random.randint(60, 100, 100),
    '类别':np.random.choice(['A班', 'B班', 'C班'], 100)

}
)
products = pd.DataFrame({ 
    '价格': np.random.exponential(50, 200), #exponential为指数分布
    '销量': np.random.poisson(20, 200), #poisson为泊松分布
    '评分': np.random.normal(4, 0.5, 200), #
    '库存': np.random.randint(0, 50, 200),
    '类型': np.random.choice(['电子', '服装', '食品'], 200), #随机分配200个类别
}
)
g1 = sns.pairplot( # 成对关系图
    data=penguins,
    hue='species',
    palette='Set2',
    vars=['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g'],
    diag_kind='kde', #对角线为KDE图
    plot_kws={'alpha':0.7,
              's':40}
)
g1.figure.suptitle(
    "示例1：企鹅数据集的Pairplot分析",
    fontsize=16,
    fontweight='bold',
    y=1.02
)
g1.tight_layout()
plt.show()

```

### FacetGrid()函数
FacetGrid()函数用于创建分面网格图，显示多个子图以展示数据的不同方面。它通过将数据按照一个或多个分类变量进行分组，并为每个组绘制一个子图，适用于探索数据的多维关系和分布特征。
#### FacetGrid(data, row, col)函数参数
- row, col：用于指定分面变量。row表示按行分面的变量，col表示按列分面的变量,如果子图数量超过列的最大数量，会自动换行。
- col_wrap：用于指定列的最大数量。当子图数量超过col_wrap时，会自动换行。
- height：用于指定每个子图的高度。它是一个正数，表示子图的高度，单位为英寸。
- aspect：用于指定子图的宽高比。它是一个正数，表示子图的宽度与高度的比例。
- sharex, sharey：用于指定子图是否共享x轴和y轴。
- margin_titles：用于指定是否在子图上方显示分面变量的标题。
- despine：用于指定是否去除子图的边框。
- hue：用于指定颜色编码的变量，可以根据某个分类变量为不同组分配不同颜色。
- palette：用于指定调色板，可以选择不同的颜色方案。
#### map(func, x, y, **kwargs)函数
- func可以为：
  - scatterplot: 散点图
  - lineplot: 折线图
  - histplot: 直方图
  - kdeplot: 核密度估计图
  - boxplot: 箱线图
  - violinplot: 小提琴图
  - barplot: 条形图
  - regplot: 回归图
  - barplot: 条形图
  - plt.scatter: matplotlib散点图, plt.plot: matplotlib折线图等
- x, y: 指定要绘制的变量
- **kwargs: 传递给绘图函数的其他参数
  - s：用于指定散点图中数据点的大小。它是一个正数，表示数据点的大小。
  - alpha：用于指定图形的透明度。它是一个介于0和1之间的浮点数，表示图形的透明度。
  - color：用于指定图形的颜色。它可以是颜色名称（如'red'、'blue'等）、十六进制颜色代码（如'#FF0000'）等。
  - marker：用于指定散点图中数据点的标记样式。它可以是一个字符串，表示数据点的标记样式（如'o'表示圆形，'^'表示三角形等）。

```python
gf1 = sns.FacetGrid(
    tips,
    col='day',
    col_wrap=2,
    height=3.5,
    aspect=1.3,
)
gf1.map(sns.violinplot, 'time', 'total_bill'
        , inner='quartile',order=['Lunch', 'Dinner']
        )
gf1.figure.suptitle(
    'Seaborn FacetGrid - 小费数据集小提琴图示例',
    fontsize=16,
    fontweight='bold',
    y=1.02
)
gf1.figure.tight_layout()
```

#### np.repeat()函数和 np.tile()函数
repeat单个重复指定次数后拼接， tile整体重复指定次数后拼接  
- np.repeat()函数用于沿指定轴重复数组的元素。它接受一个数组和一个重复次数作为输入，并返回一个新的数组，其中每个元素根据指定的重复次数进行重复。
- np.tile()函数用于沿指定轴重复整个数组。它接受一个数组和一个重复次数作为输入，并返回一个新的数组，其中整个数组根据指定的重复次数进行重复。

```python
students = pd.DataFrame({
    '成绩': np.concatenate([ #concatenate用于连接多个数组
        np.random.normal(75, 10, 100),  # A班 #正态分布，均值75，标准差10，样本量100
        np.random.normal(65, 15, 100),  # B班
        np.random.normal(55, 20, 100)   # C班
    ]),
    '科目': np.repeat(['语文', '数学', '英语'], 100), #重复每个科目100次
    '班级': np.tile(['A班', 'B班', 'C班'], 100), #tile用于重复数组，形成班级列
    '性别': np.random.choice(['男', '女'], 300) #随机分配性别
})
print(students.head(-100))
gf5 = sns.FacetGrid(
    students,
    col='科目',
    row='性别',
    height=3,
    aspect=1.3,
    hue='班级',
    palette='Set1'
)
gf5.map(
    sns.kdeplot,
    '成绩',
    fill=True,
)
gf5.add_legend(title='班级', ncol=1) #图例标题为班级，1列显示图例
gf5.set_axis_labels('成绩', '密度') #设置x轴和y轴标签
gf5.figure.suptitle(
    '示例8：学生成绩数据集的FacetGrid多重分面分析',
    fontsize=16,
    fontweight='bold',
    y=1.02
)
```

### PairGrid()函数
PairGrid是pairplot的底层实现。 PairGrid()函数用于创建成对关系网格图，显示多个变量之间的两两关系。它通过绘制散点图矩阵来展示变量之间的关系，适用于探索多个连续变量之间的相互关系。与pairplot()函数不同，PairGrid()函数提供了更高的灵活性，允许用户自定义每个子图的绘制方式和样式。
#### PairGrid(data, hue, )函数参数
- data：用于指定绘图的数据。它可以是一个DataFrame或类似的结构。
- hue：用于指定分组变量。可以将数据按照hue变量的不同取值进行分组，并为每个组绘制不同的颜色。默认情况下，hue=None，表示不进行分组。
- palette：用于指定颜色调色板。可以设置为预定义的调色板名称（如'deep'、'muted'、'bright'等），也可以设置为自定义的颜色列表。默认情况下，palette=None，表示使用默认的颜色调色板。
- diag_sharey：用于指定对角线图是否共享y轴。默认情况下，diag_sharey=True，表示对角线图共享y轴。可以将其设置为False，以使对角线图不共享y轴。
- height：用于指定每个子图的高度。它是一个正数，表示子图的高度，单位为英寸。
- aspect：用于指定子图的宽高比。它是一个正数，表示子图的宽度与高度的比例。
- despine：用于指定是否去除子图的边框。
