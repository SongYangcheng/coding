## pairplot

```python
import seaborn as sns  # 导入seaborn库，用于高级数据可视化，包括pairplot
import matplotlib.pyplot as plt  # 导入matplotlib.pyplot模块，用于创建和定制图表
import pandas as pd  # 导入pandas库，用于数据处理和创建DataFrame
import numpy as np  # 导入numpy库，用于数值计算和生成随机数据

# 设置中文字体和样式
plt.rcParams['font.family'] = ['Times New Roman', 'STsong']  # 设置全局字体，支持英文（Times New Roman）和中文（宋体）
plt.rcParams['axes.unicode_minus'] = False  # 确保负号在中文环境下正确显示

# 准备数据集
np.random.seed(42)  # 设置随机种子为42，确保生成的数据可重现

# 1. 鸢尾花数据集（经典多分类数据）
iris = sns.load_dataset('iris')  # 加载seaborn内置的iris数据集，包含花瓣/花萼尺寸和物种
# 用途：iris适合展示多变量分类关系的pairplot

# 2. 企鹅数据集
penguins = sns.load_dataset('penguins')  # 加载seaborn内置的penguins数据集，包含企鹅体征和种类
# 用途：penguins适合探索多变量和分类特征的关系

# 3. 小费数据集
tips = sns.load_dataset('tips')  # 加载seaborn内置的tips数据集，包含小费、账单和客户信息
# 用途：tips适合分析数值和分类变量的交互关系

# 4. 钻石数据集（采样）
diamonds = sns.load_dataset('diamonds').sample(500, random_state=42)  # 加载diamonds数据集，随机抽样500条，控制随机种子
# 用途：diamonds适合展示大规模数据的变量关系（价格、克拉等）

# 5. 自定义数据集 - 学生成绩
students = pd.DataFrame({
    '语文': np.random.randint(60, 100, 100),  # 生成100个60-100的随机语文成绩
    '数学': np.random.randint(60, 100, 100),  # 生成100个60-100的随机数学成绩
    '英语': np.random.randint(60, 100, 100),  # 生成100个60-100的随机英语成绩
    '物理': np.random.randint(60, 100, 100),  # 生成100个60-100的随机物理成绩
    '化学': np.random.randint(60, 100, 100),  # 生成100个60-100的随机化学成绩
    '类别': np.random.choice(['理科班', '文科班', '实验班'], 100)  # 随机分配100个类别（理科班/文科班/实验班）
})
# 用途：students适合展示分类变量（类别）与多门学科成绩的关系

# 6. 自定义数据集 - 产品数据
products = pd.DataFrame({
    '价格': np.random.exponential(50, 200),  # 生成200个指数分布的价格数据，均值50
    '销量': np.random.poisson(100, 200),  # 生成200个泊松分布的销量数据，均值100
    '评分': np.random.beta(8, 2, 200) * 5,  # 生成200个贝塔分布的评分数据，缩放至0-5
    '库存': np.random.randint(0, 500, 200),  # 生成200个0-500的随机库存数据
    '类型': np.random.choice(['电子产品', '服装', '食品'], 200)  # 随机分配200个产品类型
})
# 用途：products适合分析商业数据中数值和分类变量的模式

print("=" * 80)  # 打印80个等号，创建分隔线，提升输出可读性
print("📊 Seaborn pairplot 教程")  # 打印教程标题，表明主题为pairplot教学
print("=" * 80)  # 打印分隔线，保持格式一致
print("\n⚠️  重要提示:")  # 打印提示标题，突出pairplot特性说明
print("  pairplot 是【图级函数】(Figure-level)，用于多变量探索性分析")  # 说明pairplot为图级函数，自动管理画布
print("  • pairplot 自动创建 Figure，展示所有变量的成对关系")  # 说明pairplot生成变量对的散点图矩阵
print("  • pairplot 返回 PairGrid 对象")  # 说明pairplot返回PairGrid对象，便于进一步定制
print("  • 对角线：显示单变量分布（直方图/KDE）")  # 说明对角线展示单变量分布
print("  • 非对角线：显示两变量关系（散点图/回归图/KDE）")  # 说明非对角线展示变量间关系
print("  • 适合快速探索数据集中的关系和模式")  # 说明pairplot的探索性分析用途
print("=" * 80)  # 打印分隔线，结束提示部分
```

```python
# ==================== 第一部分：基础用法 ====================
print("\n" + "=" * 80)  # 打印换行和80个等号，创建分隔线，标记第一部分开始
print("【第一部分】基础用法")  # 打印部分标题，说明展示pairplot基础用法
print("=" * 80)  # 打印80个等号，结束分隔线

# ========== 示例 1：最基础的 pairplot ==========
print("\n# ========== 示例 1：基础 pairplot ==========")  # 打印示例标题，说明基础pairplot
print("  显示鸢尾花数据集所有数值变量的成对关系")  # 说明示例展示iris数据集变量关系
g1 = sns.pairplot(iris)  # 绘制基础pairplot，使用iris数据集（4个数值列+物种）
# 参数说明：默认对角线显示直方图，非对角线显示散点图，包含所有数值变量
g1.fig.suptitle(
    '示例1: 基础pairplot\n(对角线=直方图, 非对角线=散点图)',  # 设置标题，说明图表内容
    fontsize=14,  # 标题字体大小14，适中醒目
    fontweight='bold',  # 标题粗体，增强视觉效果
    y=1.01)  # 标题位置略高于画布顶部，避免遮挡
plt.tight_layout()  # 自动调整图表间距，优化布局
plt.show()  # 显示图表，渲染基础pairplot

# ========== 示例 2：hue参数 - 按类别着色 ==========
print("\n# ========== 示例 2：hue参数 - 按类别着色 ==========")  # 打印示例标题，说明按类别着色
print("  不同鸢尾花种类用不同颜色显示")  # 说明示例按物种着色
g2 = sns.pairplot(iris, hue='species', palette='Set2')  # 绘制pairplot，按物种着色
# 参数说明：hue='species'按物种分类着色，palette='Set2'使用柔和颜色集
g2.fig.suptitle(
    '示例2: hue="species"\n(按品种分组着色)',  # 设置标题，说明hue参数效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染带类别颜色的pairplot

# ========== 示例 3：markers参数 - 不同标记样式 ==========
print("\n# ========== 示例 3：markers参数 - 不同标记 ==========")  # 打印示例标题，说明不同标记样式
print("  不同类别使用不同的标记符号")  # 说明示例为不同物种设置不同标记
g3 = sns.pairplot(iris, hue='species', markers=['o', 's', 'D'],
                  palette='husl')  # 绘制pairplot，使用不同标记
# 参数说明：hue='species'按物种着色，markers=['o', 's', 'D']为三种物种设置圆形、方形、菱形，palette='husl'使用均匀色调
g3.fig.suptitle(
    '示例3: markers参数\n(圆形/方形/菱形)',  # 设置标题，说明标记样式效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染带不同标记的pairplot

# ========== 示例 4：diag_kind参数 - 对角线KDE ==========
print("\n# ========== 示例 4：diag_kind='kde' - 对角线核密度估计 =========="
      )  # 打印示例标题，说明对角线KDE
print("  对角线显示平滑的密度曲线而非直方图")  # 说明示例用KDE替代直方图
g4 = sns.pairplot(iris, hue='species', diag_kind='kde',
                  palette='deep')  # 绘制pairplot，使用KDE对角线
# 参数说明：hue='species'按物种着色，diag_kind='kde'对角线显示核密度估计，palette='deep'使用深色调
g4.fig.suptitle(
    '示例4: diag_kind="kde"\n(对角线显示密度曲线)',  # 设置标题，说明KDE效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染带KDE对角线的pairplot

# ========== 示例 5：kind参数 - 回归图 ==========
print("\n# ========== 示例 5：kind='reg' - 非对角线显示回归线 ==========")  # 打印示例标题，说明回归图
print("  在散点图上添加回归拟合线")  # 说明示例在散点图添加回归线
g5 = sns.pairplot(iris, hue='species', kind='reg',
                  palette='colorblind')  # 绘制pairplot，添加回归线
# 参数说明：hue='species'按物种着色，kind='reg'非对角线显示散点图+回归线，palette='colorblind'使用色盲友好颜色
g5.fig.suptitle(
    '示例5: kind="reg"\n(散点图 + 回归拟合线)',  # 设置标题，说明回归线效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染带回归线的pairplot
```

```python
# ==================== 第二部分：变量选择与布局控制 ====================
print("\n" + "=" * 80)  # 打印换行和80个等号，创建分隔线，标记第二部分开始
print("【第二部分】变量选择与布局控制")  # 打印部分标题，说明展示pairplot变量选择与布局控制
print("=" * 80)  # 打印80个等号，结束分隔线

# ========== 示例 6：vars参数 - 选择特定变量 ==========
print("\n# ========== 示例 6：vars参数 - 选择子集变量 ==========")  # 打印示例标题，说明选择特定变量
print("  只显示指定的几个变量")  # 说明示例仅展示部分变量
g6 = sns.pairplot(
    iris,  # 使用iris数据集
    vars=['sepal_length', 'sepal_width', 'petal_length'],  # 仅选择花萼长度、宽度和花瓣长度
    hue='species',  # 按物种着色
    palette='Set1')  # 使用Set1颜色集，鲜艳区分
# 参数说明：vars指定绘制的变量子集，生成3x3矩阵，hue='species'按物种分类着色
g6.fig.suptitle(
    '示例6: vars参数\n(只显示3个变量)',  # 设置标题，说明vars参数效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距，优化布局
plt.show()  # 显示图表，渲染仅含3个变量的pairplot

# ========== 示例 7：x_vars和y_vars - 非对称矩阵 ==========
print("\n# ========== 示例 7：x_vars和y_vars - 非对称布局 ==========")  # 打印示例标题，说明非对称布局
print("  X轴和Y轴使用不同的变量集")  # 说明示例使用不同变量集生成非对称矩阵
g7 = sns.pairplot(
    iris,  # 使用iris数据集
    x_vars=['sepal_length', 'sepal_width'],  # x轴变量：花萼长度、宽度
    y_vars=['petal_length', 'petal_width'],  # y轴变量：花瓣长度、宽度
    hue='species',  # 按物种着色
    palette='viridis',  # 使用viridis颜色集，渐变效果
    height=3)  # 设置每个子图高度为3英寸
# 参数说明：x_vars和y_vars生成2x2非对称矩阵，hue='species'按物种分类着色
g7.fig.suptitle(
    '示例7: x_vars和y_vars\n(2x2非对称矩阵)',  # 设置标题，说明非对称矩阵效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染非对称pairplot

# ========== 示例 8：corner参数 - 只显示下三角 ==========
print("\n# ========== 示例 8：corner=True - 下三角矩阵 ==========")  # 打印示例标题，说明下三角矩阵
print("  避免重复，只显示下三角部分")  # 说明示例仅显示下三角，避免重复
g8 = sns.pairplot(iris, hue='species', corner=True,
                  palette='muted')  # 绘制pairplot，仅显示下三角
# 参数说明：corner=True仅绘制下三角矩阵（去除上三角和对角线镜像），hue='species'按物种着色，palette='muted'使用柔和颜色
g8.fig.suptitle(
    '示例8: corner=True\n(只显示下三角，避免重复)',  # 设置标题，说明下三角效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染下三角pairplot

# ========== 示例 9：height和aspect - 调整图形大小 ==========
print("\n# ========== 示例 9：height和aspect - 控制大小 ==========")  # 打印示例标题，说明图形大小调整
print("  调整每个子图的高度和宽高比")  # 说明示例调整子图尺寸
g9 = sns.pairplot(
    penguins.dropna(),  # 使用penguins数据集，去除缺失值
    hue='species',  # 按物种着色
    height=2,  # 设置每个子图高度为2英寸
    aspect=1.5,  # 设置子图宽高比为1.5（较宽）
    palette='Set2')  # 使用Set2颜色集，柔和区分
# 参数说明：dropna()去除缺失值，height=2和aspect=1.5调整子图大小，生成更宽的子图
g9.fig.suptitle(
    '示例9: height=2, aspect=1.5\n(较小较宽的子图)',  # 设置标题，说明尺寸调整效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染调整尺寸的pairplot
```

```python
# ==================== 第三部分：高级定制 ====================
print("\n" + "=" * 80)  # 打印换行和80个等号，创建分隔线，标记第三部分开始
print("【第三部分】高级定制")  # 打印部分标题，说明展示pairplot高级定制
print("=" * 80)  # 打印80个等号，结束分隔线

# ========== 示例 10：plot_kws - 散点图样式 ==========
print("\n# ========== 示例 10：plot_kws - 自定义散点图 ==========")  # 打印示例标题，说明散点图样式定制
print("  调整散点的大小、透明度等")  # 说明示例调整散点图样式
g10 = sns.pairplot(
    iris,
    hue='species',  # 使用iris数据集，按物种着色
    plot_kws={
        'alpha': 0.6,
        's': 80,
        'edgecolor': 'black'
    },  # 自定义散点图：透明度0.6，点大小80，黑色边框
    diag_kind='kde',  # 对角线使用核密度估计（KDE）
    palette='bright')  # 使用明亮颜色集
# 参数说明：plot_kws控制非对角线散点图样式，alpha设置透明度，s设置点大小，edgecolor设置边框颜色
g10.fig.suptitle(
    '示例10: plot_kws\n(半透明 + 黑色边框 + 大散点)',  # 设置标题，说明散点图样式效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距，优化布局
plt.show()  # 显示图表，渲染自定义散点图样式的pairplot

# ========== 示例 11：diag_kws - 对角线样式 ==========
print("\n# ========== 示例 11：diag_kws - 自定义对角线 ==========")  # 打印示例标题，说明对角线样式定制
print("  调整直方图或KDE的样式")  # 说明示例调整对角线直方图样式
g11 = sns.pairplot(
    iris,
    hue='species',
    diag_kind='hist',  # 使用iris数据集，按物种着色，对角线使用直方图
    diag_kws={
        'bins': 20,
        'alpha': 0.7,
        'edgecolor': 'black'
    },  # 自定义直方图：20个bins，透明度0.7，黑色边框
    palette='pastel')  # 使用柔和颜色集
# 参数说明：diag_kws控制对角线直方图样式，bins设置柱数，alpha设置透明度，edgecolor设置边框颜色
g11.fig.suptitle(
    '示例11: diag_kws\n(20个bins + 半透明直方图)',  # 设置标题，说明直方图样式效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染自定义直方图样式的pairplot

# ========== 示例 12：kind='kde' - KDE密度图 ==========
print("\n# ========== 示例 12：kind='kde' - 双变量密度图 ==========")  # 打印示例标题，说明双变量KDE
print("  非对角线显示2D核密度估计")  # 说明示例用KDE显示双变量关系
g12 = sns.pairplot(
    iris,
    hue='species',
    kind='kde',  # 使用iris数据集，按物种着色，非对角线使用2D KDE
    diag_kind='kde',  # 对角线使用1D KDE
    palette='rocket')  # 使用rocket颜色集，渐变效果
# 参数说明：kind='kde'设置非对角线为2D核密度估计图，diag_kind='kde'保持对角线为1D KDE
g12.fig.suptitle(
    '示例12: kind="kde"\n(全部使用密度估计)',  # 设置标题，说明KDE效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染全KDE的pairplot

# ========== 示例 13：palette - 自定义调色板 ==========
print("\n# ========== 示例 13：palette - 多种调色板 ==========")  # 打印示例标题，说明自定义调色板
print("  使用自定义颜色字典")  # 说明示例使用自定义颜色字典
custom_palette = {
    'setosa': '#FF6B6B',
    'versicolor': '#4ECDC4',
    'virginica': '#45B7D1'
}  # 定义物种颜色字典
g13 = sns.pairplot(
    iris,
    hue='species',
    palette=custom_palette,  # 使用iris数据集，按物种着色，自定义调色板
    diag_kind='kde')  # 对角线使用KDE
# 参数说明：palette使用自定义十六进制颜色字典，为每种物种指定独特颜色
g13.fig.suptitle(
    '示例13: palette自定义\n(自定义颜色)',  # 设置标题，说明自定义调色板效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染自定义调色板的pairplot

# ========== 示例 14：多类别数据 ==========
print("\n# ========== 示例 14：多类别数据（企鹅数据集） ==========")  # 打印示例标题，说明多类别数据
print("  展示更复杂的多类别数据关系")  # 说明示例展示企鹅数据集的多变量关系
g14 = sns.pairplot(
    penguins.dropna(),
    hue='species',  # 使用penguins数据集，去除缺失值，按物种着色
    vars=[
        'bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g'
    ],  # 选择4个数值变量
    diag_kind='kde',  # 对角线使用KDE
    palette='Set1',  # 使用Set1颜色集，鲜艳区分
    plot_kws={
        'alpha': 0.6,
        's': 50
    })  # 自定义散点图：透明度0.6，点大小50
# 参数说明：dropna()去除缺失值，vars选择特定变量，plot_kws调整散点图样式
g14.fig.suptitle(
    '示例14: 企鹅数据集\n(4个测量指标 × 3个物种)',  # 设置标题，说明企鹅数据集效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染企鹅数据集的pairplot
```

```python
# ========== 案例 6：综合应用 - 完整配置 ==========
print("\n# ========== 案例 6：综合应用 - 完整参数配置 ==========")  # 打印案例标题，说明综合应用案例
print("  展示pairplot的完整功能")  # 说明案例目标：展示pairplot的多参数配置
# 创建综合数据集
comprehensive_data = pd.DataFrame({
    '指标A':
    np.random.randn(150) * 10 + 50,  # 生成150个正态分布数据，均值50，标准差10
    '指标B':
    np.random.randn(150) * 15 + 60,  # 生成150个正态分布数据，均值60，标准差15
    '指标C':
    np.random.randn(150) * 8 + 70,  # 生成150个正态分布数据，均值70，标准差8
    '指标D':
    np.random.randn(150) * 12 + 55,  # 生成150个正态分布数据，均值55，标准差12
    '分组':
    np.random.choice(['优秀', '良好', '及格'], 150,
                     p=[0.3, 0.5, 0.2])  # 随机分配分组，概率分别为0.3、0.5、0.2
})
# 用途：comprehensive_data模拟多变量数据，适合展示pairplot的全面功能

g20 = sns.pairplot(
    comprehensive_data,  # 使用综合数据集
    hue='分组',  # 按分组（优秀/良好/及格）着色
    vars=['指标A', '指标B', '指标C', '指标D'],  # 选择4个指标变量
    diag_kind='kde',  # 对角线使用核密度估计（KDE）
    kind='scatter',  # 非对角线使用散点图
    palette='Set1',  # 使用Set1颜色集，鲜艳区分类别
    markers=['o', 's', 'D'],  # 为优秀、良好、及格设置圆形、方形、菱形标记
    height=2.5,  # 设置每个子图高度为2.5英寸
    aspect=1,  # 设置子图宽高比为1（正方形）
    corner=False,  # 显示完整矩阵，包含上三角
    plot_kws={  # 自定义散点图样式
        'alpha': 0.6,  # 设置散点透明度为0.6
        's': 50,  # 设置散点大小为50
        'edgecolor': 'white',  # 设置散点边框为白色
        'linewidth': 0.5  # 设置边框线宽为0.5
    },
    diag_kws={  # 自定义对角线KDE样式
        'alpha': 0.7,  # 设置KDE透明度为0.7
        'linewidth': 2  # 设置KDE线条宽度为2
    })
# 参数说明：全面配置参数，结合hue、vars、markers、plot_kws、diag_kws等，生成专业级pairplot
g20.fig.suptitle(
    '案例6: 综合应用\n(完整参数配置 + 精细美化)',  # 设置标题，说明综合配置效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高

# 添加图例位置调整
g20._legend.set_bbox_to_anchor((1.05, 0.5))  # 将图例移到右侧，垂直居中
g20._legend.set_title('数据分组', prop={
    'size': 11,
    'weight': 'bold'
})  # 设置图例标题为“数据分组”，字体大小11，粗体

plt.tight_layout()  # 自动调整图表间距，优化布局
plt.show()  # 显示图表，渲染综合配置的pairplot
```

```python
# ==================== 第一部分：基础用法 ====================
print("\n" + "=" * 80)  # 打印换行和80个等号，创建分隔线，标记第一部分开始
print("【第一部分】基础用法 - 单维度分面")  # 打印部分标题，说明展示FacetGrid单维度分面用法
print("=" * 80)  # 打印80个等号，结束分隔线

# ========== 示例 1：col参数 - 按列分面 ==========
print("\n# ========== 示例 1：col参数 - 按列分面 ==========")  # 打印示例标题，说明按列分面
print("  每个列显示一个时间段的数据")  # 说明示例按用餐时间分列展示数据
g1 = sns.FacetGrid(tips, col='time', height=4,
                   aspect=1.2)  # 创建FacetGrid，按time（午餐/晚餐）分列
# 参数说明：tips数据集，col='time'按时间列分面，height=4设置子图高度4英寸，aspect=1.2设置宽高比
g1.map(sns.scatterplot, 'total_bill', 'tip', alpha=0.6,
       s=50)  # 映射散点图，x轴为total_bill，y轴为tip
# 参数说明：map使用sns.scatterplot绘制散点图，alpha=0.6设置透明度，s=50设置点大小
g1.set_axis_labels('总账单 ($)', '小费 ($)')  # 设置x轴标签为“总账单 ($)”，y轴标签为“小费 ($)”
g1.fig.suptitle(
    '示例1: col="time"\n(按用餐时间分列)',  # 设置标题，说明按时间分列效果
    fontsize=14,
    fontweight='bold',
    y=1.02)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距，优化布局
plt.show()  # 显示图表，渲染按用餐时间分列的FacetGrid散点图
```

```python
# ========== 示例 2：row参数 - 按行分面 ==========
print("\n# ========== 示例 2：row参数 - 按行分面 ==========")  # 打印示例标题，说明按行分面
print("  每一行显示一个吸烟状态")  # 说明示例按吸烟状态分行展示数据
g2 = sns.FacetGrid(tips, row='smoker', height=3,
                   aspect=2)  # 创建FacetGrid，按smoker（吸烟/非吸烟）分行
# 参数说明：tips数据集，row='smoker'按吸烟状态分行，height=3设置子图高度3英寸，aspect=2设置宽高比
g2.map(sns.histplot, 'total_bill', bins=15, kde=True)  # 映射直方图，显示total_bill分布
# 参数说明：map使用sns.histplot绘制直方图，bins=15设置15个柱，kde=True添加核密度曲线
g2.set_axis_labels('总账单 ($)', '频数')  # 设置x轴标签为“总账单 ($)”，y轴标签为“频数”
g2.fig.suptitle(
    '示例2: row="smoker"\n(按吸烟状态分行)',  # 设置标题，说明按吸烟状态分行效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距，优化布局
plt.show()  # 显示图表，渲染按吸烟状态分行的直方图
```

```python
# ========== 示例 3：hue参数 - 按颜色分组 ==========
print("\n# ========== 示例 3：hue参数 - 按颜色分组 ==========")  # 打印示例标题，说明按颜色分组
print("  用不同颜色区分性别")  # 说明示例按性别着色
g3 = sns.FacetGrid(tips, hue='sex', height=5, aspect=1.5,
                   palette='Set1')  # 创建FacetGrid，按sex（性别）着色
# 参数说明：tips数据集，hue='sex'按性别着色，height=5设置子图高度5英寸，aspect=1.5设置宽高比，palette='Set1'使用鲜艳颜色集
g3.map(sns.scatterplot, 'total_bill', 'tip', alpha=0.6,
       s=60)  # 映射散点图，显示total_bill与tip关系
# 参数说明：map使用sns.scatterplot绘制散点图，alpha=0.6设置透明度，s=60设置点大小
g3.add_legend(title='性别')  # 添加图例，标题为“性别”
g3.set_axis_labels('总账单 ($)', '小费 ($)')  # 设置x轴标签为“总账单 ($)”，y轴标签为“小费 ($)”
g3.fig.suptitle(
    '示例3: hue="sex"\n(按性别着色)',  # 设置标题，说明按性别着色效果
    fontsize=14,
    fontweight='bold',
    y=1.02)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染按性别着色的散点图
```

```python
# ========== 示例 4：col_wrap参数 - 自动换行 ==========
print("\n# ========== 示例 4：col_wrap - 控制每行列数 ==========")  # 打印示例标题，说明自动换行
print("  将多个分面自动换行显示")  # 说明示例按天分列并自动换行
g4 = sns.FacetGrid(tips, col='day', col_wrap=2, height=3.5,
                   aspect=1.3)  # 创建FacetGrid，按day（星期）分列，每行2列
# 参数说明：tips数据集，col='day'按星期分列，col_wrap=2每行显示2个子图，height=3.5设置子图高度，aspect=1.3设置宽高比
g4.map(sns.violinplot, 'time', 'total_bill',
       order=['Lunch', 'Dinner'])  # 映射小提琴图，显示time与total_bill关系
# 参数说明：map使用sns.violinplot绘制小提琴图，time为x轴，total_bill为y轴，order指定时间顺序
g4.set_axis_labels('用餐时间', '总账单 ($)')  # 设置x轴标签为“用餐时间”，y轴标签为“总账单 ($)”
g4.fig.suptitle(
    '示例4: col_wrap=2\n(4天数据，每行2个)',  # 设置标题，说明换行效果
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题字体14，粗体，位置略高
plt.tight_layout()  # 自动调整图表间距
plt.show()  # 显示图表，渲染按星期分列并换行的小提琴图
```

## PairGrid

```python
"""
================================================================================
                Seaborn PairGrid 完整教程
          灵活的成对关系网格系统 - 底层自定义工具
================================================================================

教程说明:
PairGrid 是 pairplot 的底层实现，提供了更强大的自定义能力。

PairGrid vs pairplot:
• pairplot: 高级函数，自动创建成对关系图，使用简单
• PairGrid: 底层网格类，可以完全自定义每个位置的图表

核心优势:
• 对角线可以绘制不同的图（直方图、KDE、箱线图等）
• 上下三角可以绘制不同的图（散点图、回归图、KDE等）
• 可以使用自定义函数
• 更精细的控制和更灵活的组合

适合场景:
• 需要精细控制每个子图的内容
• 需要在不同区域显示不同类型的图表
• 需要使用自定义绘图函数
================================================================================
"""

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
import warnings
warnings.filterwarnings('ignore')


print("=" * 80)
print("              📊 Seaborn PairGrid 完整教程")
print("=" * 80)

# ==================== 数据准备 ====================
print("\n【数据准备】加载和生成数据集...")

# 1. 鸢尾花数据集（经典）
iris = pd.read_csv('./data/seaborn-data-master/iris.csv')

# 2. 企鹅数据集
penguins = pd.read_csv('./data/seaborn-data-master/penguins.csv')

# 3. 小费数据集
tips = pd.read_csv('./data/seaborn-data-master/tips.csv')

# 4. 自定义多维数据集
np.random.seed(42)
n_samples = 200
custom_data = pd.DataFrame({
    '变量A': np.random.normal(50, 10, n_samples),
    '变量B': np.random.normal(60, 15, n_samples),
    '变量C': np.random.normal(55, 12, n_samples),
    '变量D': np.random.normal(70, 8, n_samples),
    '类别': np.random.choice(['类型1', '类型2', '类型3'], n_samples)
})

# 添加一些相关性
custom_data['变量B'] = custom_data['变量A'] * 0.7 + np.random.normal(20, 5, n_samples)
custom_data['变量C'] = custom_data['变量A'] * 0.3 + custom_data['变量B'] * 0.4 + np.random.normal(10, 8, n_samples)

print("✅ 数据准备完成!")
print(f"  • 鸢尾花数据: {iris.shape}")
print(f"  • 企鹅数据: {penguins.shape}")
print(f"  • 小费数据: {tips.shape}")
print(f"  • 自定义数据: {custom_data.shape}")
```

```python
# ==================== PairGrid vs pairplot 对比 ====================  # 章节标题注释：本段展示 Seaborn 高级函数 pairplot 与底层类 PairGrid 的用法与差异
print("\n" + "=" * 80)  # 打印分隔线：80 个等号，前置换行便于阅读
print("【对比演示】PairGrid vs pairplot")  # 打印当前演示主题
print("=" * 80)  # 再打印一条分隔线，形成上下呼应的视觉区块

print("\n示例1: 使用 pairplot（高级函数）")  # 说明：首先展示更高抽象层的便捷接口 pairplot
g_pairplot = sns.pairplot(
    iris, hue='species',
    height=2.5)  # 自动生成数值变量两两组合图；hue 根据类别着色；height 每个子图高度（英寸）
g_pairplot.fig.suptitle(
    'pairplot: 自动创建（简单但不够灵活）',  # 设置整张图的总标题：强调“自动创建”的特点
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 标题样式与位置：稍微上移避免与子图重叠
plt.tight_layout()  # 自动调整子图间距与边距，减少标签与标题的重叠
plt.show()  # 渲染显示：输出成对关系的散点图与对角分布图（默认对角线是直方图/核密度，依版本而异）

print("\n示例2: 使用 PairGrid（底层类）- 实现相同效果")  # 说明：演示底层类，可手工指定各区域映射
g_pairgrid = sns.PairGrid(iris, hue='species',
                          height=2.5)  # 构建配对网格骨架：仅创建网格与轴对象，不绘图；参数与 pairplot 类似
g_pairgrid.map_diag(sns.histplot)  # 对角线：映射为直方图（每个变量自身的分布），可自由改为 kdeplot 等
g_pairgrid.map_offdiag(sns.scatterplot)  # 非对角线：映射为散点图（变量两两关系），比 pairplot 更可控
g_pairgrid.add_legend()  # 添加图例：显示 hue 类别对应的颜色/标记，提升可读性
g_pairgrid.fig.suptitle(
    'PairGrid: 手动构建（灵活可定制）',  # 设置整图标题：强调“手动构建”“灵活可定制”
    fontsize=14,
    fontweight='bold',
    y=1.01)  # 字体样式与位置设置
plt.tight_layout()  # 调整布局：在添加 suptitle 与图例后，进一步优化留白
plt.show()  # 渲染显示：与你的 pairplot 视觉近似，但每个子区的图类型可独立定制

print("\n💡 关键区别:")  # 小结引导：突出对比要点
print("  pairplot: g = sns.pairplot(data, hue='species')"
      )  # pairplot 一行生成：高层封装，适合快速探索
print("  PairGrid: g = sns.PairGrid(data, hue='species')"
      )  # PairGrid 先创建网格对象，再逐步映射
print("           g.map_diag(sns.histplot)")  # 指定对角线图型：自由度高
print("           g.map_offdiag(sns.scatterplot)")  # 指定非对角线图型：可替换为回归线、密度图等
print("           g.add_legend()")  # 手动添加图例：步骤式更清晰，教学展示便于讲解控制流程
```

```python
# ==================== 第一部分：基础用法 ====================  # 章节标题注释：本段演示 PairGrid 的基础图形用法与参数
print("\n" + "=" * 80)  # 打印 80 个“=”作为分隔线；前置换行让输出更清晰
print("【第一部分】PairGrid 基础用法")  # 打印段落标题，便于在控制台阅读
print("=" * 80)  # 再打印一条分隔线，形成上/下呼应的视觉区块

# ========== 示例3: 最基础的 PairGrid ==========  # 示例3 小标题：统一在所有单元上绘制同一种图
print("\n# ========== 示例3: 基础 PairGrid ==========")  # 打印该示例的标题（带前置换行）
g1 = sns.PairGrid(iris)  # 构建 PairGrid 网格：默认对 iris 的所有数值列做两两组合
g1.map(sns.scatterplot)  # 所有位置都画散点图  # 在整个网格（对角线与非对角线）统一映射为散点图
g1.fig.suptitle(
    '示例3: 基础PairGrid\n(所有位置统一使用散点图)',  # 设置整图标题（含换行说明）
    fontsize=14,  # 标题字号
    fontweight='bold',  # 标题加粗
    y=1.01)  # 标题上移一点，避免与子图重叠
plt.tight_layout()  # 自动调整子图与边距，减少标签/标题重叠
plt.show()  # 渲染显示：得到一个所有单元均为散点图的配对矩阵

# ========== 示例4: 分区域绘图 - 对角线 vs 非对角线 ==========  # 示例4 小标题：对角线与非对角线分别指定图形
print("\n# ========== 示例4: 分区域绘图 ==========")  # 打印该示例的标题
g2 = sns.PairGrid(iris)  # 创建网格骨架：默认使用所有数值列
g2.map_diag(sns.histplot)  # 对角线：直方图  # 每个变量自身分布用直方图展示，直观对比偏度/离散度
g2.map_offdiag(sns.scatterplot)  # 非对角线：散点图  # 变量两两关系用散点图，观察线性/聚类/离群
g2.fig.suptitle(
    '示例4: 分区域绘图\n(对角线=直方图, 非对角线=散点图)',  # 整体标题与注解
    fontsize=14,
    fontweight='bold',
    y=1.01)
plt.tight_layout()  # 布局微调
plt.show()  # 渲染显示：对角看分布，非对角看相关

# ========== 示例5: 上下三角不同 ==========  # 示例5 小标题：上三角与下三角用不同图表
print("\n# ========== 示例5: 上下三角不同图表 ==========")  # 打印该示例的标题
g3 = sns.PairGrid(iris)  # 创建网格
g3.map_upper(sns.scatterplot)  # 上三角：散点图  # 直观观察变量两两关系与分布形态
g3.map_lower(sns.kdeplot)  # 下三角：KDE密度图  # 用核密度等高线/填充展示二维密度（聚类/相关的平滑视图）
g3.map_diag(sns.histplot)  # 对角线：直方图  # 自身分布；辅助判断是否正态、是否偏态
g3.fig.suptitle(
    '示例5: 上下三角不同\n(上=散点图, 下=KDE, 对角线=直方图)',  # 整体标题与说明
    fontsize=14,
    fontweight='bold',
    y=1.01)
plt.tight_layout()  # 自动留白
plt.show()  # 渲染显示：上/下三角各司其职，信息互补

# ========== 示例6: 添加hue分组 ==========  # 示例6 小标题：引入 hue 分类着色
print("\n# ========== 示例6: hue分组着色 ==========")  # 打印该示例的标题
g4 = sns.PairGrid(iris, hue='species',
                  palette='Set2')  # 指定 hue=species 进行分类着色；palette 设置调色板
g4.map_diag(sns.histplot)  # 对角线：直方图  # 按类别叠加/并列绘制，有助于对比分布差异
g4.map_offdiag(sns.scatterplot, alpha=0.6,
               s=40)  # 非对角线：散点图  # alpha 半透明避免点遮挡；s 控制点大小
g4.add_legend(title='鸢尾花品种', bbox_to_anchor=(1.05, 0.5),
              loc='center left')  # 添加图例并放到右侧，避免压住子图
g4.fig.suptitle('示例6: hue分组\n(按品种着色)', fontsize=14, fontweight='bold',
                y=1.01)  # 整体标题
plt.tight_layout()  # 调整布局：配合右侧图例留出空间
plt.show()  # 渲染显示：不同品种颜色区分，易于观察类间差异

# ========== 示例7: 选择特定变量 ==========  # 示例7 小标题：只展示部分变量
print("\n# ========== 示例7: vars参数 - 选择变量 ==========")  # 打印该示例的标题
g5 = sns.PairGrid(
    iris,
    vars=['sepal_length', 'sepal_width',
          'petal_length'],  # 只选择 3 个变量，减少维度、聚焦讲解
    hue='species')  # 继续按类别着色
g5.map_diag(sns.kdeplot, linewidth=2)  # 对角线：KDE 曲线  # 平滑展示单变量分布；加粗线条增强可读性
g5.map_offdiag(sns.scatterplot, alpha=0.6, s=50)  # 非对角线：散点图  # 更大点径便于观察
g5.add_legend()  # 自动生成图例
g5.fig.suptitle(
    '示例7: 选择特定变量\n(只显示3个变量)',  # 整体标题与说明
    fontsize=14,
    fontweight='bold',
    y=1.01)
plt.tight_layout()  # 布局微调
plt.show()  # 渲染显示：仅 3×3 子图矩阵，更利于教学聚焦

# ========== 示例8: x_vars 和 y_vars（非对称）==========  # 示例8 小标题：自定义 X 与 Y 变量形成非对称矩阵
print("\n# ========== 示例8: 非对称矩阵 ==========")  # 打印该示例的标题
g6 = sns.PairGrid(
    iris,
    x_vars=['sepal_length', 'sepal_width'],  # 作为列变量（X 轴）
    y_vars=['petal_length', 'petal_width'],  # 作为行变量（Y 轴）
    hue='species',  # 按品种着色
    height=4)  # 单个子图高度放大，便于细节观察
g6.map(sns.scatterplot, alpha=0.6, s=50)  # 在非对称的 2×2 网格上统一绘制散点图；半透明减少遮挡
g6.add_legend()  # 添加图例
g6.fig.suptitle(
    '示例8: 非对称矩阵\n(2×2 X变量 vs Y变量)',  # 整体标题与说明
    fontsize=14,
    fontweight='bold',
    y=1.02)  # 标题稍微更高，避免与图形元素重叠
plt.tight_layout()  # 自动调整边距
plt.show()  # 渲染显示：便于做“某几列 vs 另几列”的定向对比

# ==================== 第二部分：高级定制 ====================  # 章节标题注释：进入更灵活的定制技巧
print("\n" + "=" * 80)  # 分隔线：与上半部分结构对称
print("【第二部分】高级定制技巧")  # 打印第二部分标题
print("=" * 80)  # 结束分隔线

# ========== 示例9: 不同类型图表组合 ==========  # 示例9 小标题：上/下/对角使用不同图形
print("\n# ========== 示例9: 多种图表组合 ==========")  # 打印该示例的标题
g7 = sns.PairGrid(penguins, hue='species',
                  height=2.5)  # 使用企鹅数据；按 species 着色；height 控制子图尺寸
g7.map_upper(sns.scatterplot, s=30, alpha=0.6)  # 上：散点图  # 观察变量关系与类别分布
g7.map_lower(sns.kdeplot)  # 下：2D密度图  # 显示等高线或填充密度，揭示聚集区域
g7.map_diag(sns.kdeplot, linewidth=2, fill=True,
            alpha=0.5)  # 对角线：KDE  # 单变量平滑分布，fill=True 填充面积
g7.add_legend(title='企鹅品种', bbox_to_anchor=(1.05, 0.5),
              loc='center left')  # 右侧外置图例，避免遮挡
g7.fig.suptitle(
    '示例9: 多种图表组合\n(散点图 + KDE + 密度图)',  # 整体标题与说明
    fontsize=14,
    fontweight='bold',
    y=1.01)
plt.tight_layout()  # 布局优化：给右侧图例与标题留白
plt.show()  # 渲染显示：组合视图帮助从多角度理解分布与相关

# ========== 示例10: 添加回归线 ==========  # 示例10 小标题：在下三角叠加回归拟合
print("\n# ========== 示例10: 上三角散点 + 下三角回归 ==========")  # 打印该示例的标题
g8 = sns.PairGrid(iris, hue='species', height=2.5)  # 仍用 iris；按 species 着色
g8.map_upper(sns.scatterplot, s=40, alpha=0.6)  # 上三角：散点图  # 点径更大以提升辨识度
g8.map_lower(sns.regplot, scatter_kws={
    'alpha': 0.3,
    's': 20
})  # 下三角：回归图  # 带回归线与置信区间；scatter_kws 控制散点透明度与尺寸
g8.map_diag(sns.histplot, kde=True, alpha=0.7)  # 对角线：直方图 + KDE  # 同时看频数与平滑分布
g8.add_legend()  # 添加图例
g8.fig.suptitle(
    '示例10: 回归分析\n(上=散点, 下=回归线)',  # 整体标题与说明
    fontsize=14,
    fontweight='bold',
    y=1.01)
plt.tight_layout()  # 自动布局
plt.show()  # 渲染显示：便于观察线性趋势与类间差异

# ========== 示例11: 使用自定义函数 ==========  # 示例11 小标题：在网格单元中调用自定义绘制逻辑
print("\n# ========== 示例11: 自定义绘图函数 ==========")  # 打印该示例的标题


def corrfunc(x, y, **kwargs):  # 自定义函数：在单元格中计算并展示皮尔逊相关与 p 值
    """自定义函数：显示相关系数"""  # 文档字符串：提示功能
    ax = plt.gca()  # 获取当前子图轴对象，以便在其上绘制注释
    # 计算相关系数  # 对传入的 x、y 序列计算 Pearson 相关与显著性
    r, p = stats.pearsonr(x, y)  # r 为相关系数，p 为显著性水平（越小越显著）
    # 在图中显示  # 将 r 和 p 用文本框展示在单元格中心
    ax.annotate(
        f'r = {r:.2f}\np = {p:.3f}',  # f-string 控制数值格式：r 保留 2 位小数；p 保留 3 位
        xy=(0.5, 0.5),  # 注释位置位于坐标轴的中心（相对坐标）
        xycoords='axes fraction',  # 坐标系使用轴域的比例坐标
        ha='center',  # 水平居中
        va='center',  # 垂直居中
        fontsize=12,  # 字号
        bbox=dict(boxstyle='round', facecolor='wheat',
                  alpha=0.5))  # 背景框样式：圆角、浅棕底、半透明


g9 = sns.PairGrid(iris, height=2.5)  # 构建网格
g9.map_upper(sns.scatterplot, s=30, alpha=0.5)  # 上三角：半透明散点，避免遮挡
g9.map_lower(corrfunc)  # 下三角显示相关系数  # 调用自定义函数而非内置图形，增强可解释性
g9.map_diag(sns.histplot, kde=True)  # 对角线：直方 + KDE  # 同看频数与平滑曲线
g9.fig.suptitle(
    '示例11: 自定义函数\n(下三角显示相关系数)',  # 整体标题与说明
    fontsize=14,
    fontweight='bold',
    y=1.01)
plt.tight_layout()  # 布局优化
plt.show()  # 渲染显示：信息密度更高，兼顾可视与统计显著性

# ========== 示例12: 箱线图在对角线 ==========  # 示例12 小标题：对角线用箱线图以突出分组差异
print("\n# ========== 示例12: 对角线使用箱线图 ==========")  # 打印该示例的标题
g10 = sns.PairGrid(tips, hue='time', height=2.5,
                   palette='Set1')  # 使用小费数据；按就餐时间（午/晚）着色
g10.map_diag(sns.boxplot)  # 对角线：箱线图（显示分布差异）  # 观测中位数、四分位与异常值
g10.map_offdiag(sns.scatterplot, alpha=0.5, s=30)  # 非对角线：散点图  # 对比变量关系及分组分布
g10.add_legend(title='用餐时间')  # 添加图例并命名
g10.fig.suptitle(
    '示例12: 对角线箱线图\n(展示分组差异)',  # 整体标题与说明
    fontsize=14,
    fontweight='bold',
    y=1.01)
plt.tight_layout()  # 调整布局
plt.show()  # 渲染显示：分组差异在对角线箱线图上更直观

# ========== 示例13: 小提琴图组合 ==========  # 示例13 小标题：对角线用小提琴图，非对角线散点
print("\n# ========== 示例13: 小提琴图 + 散点图 ==========")  # 打印该示例的标题
g11 = sns.PairGrid(
    tips,
    hue='sex',  # 按性别分组着色
    height=2.5,  # 子图高度
    palette='husl',  # 采用 HUSL 调色盘，色相均匀
    vars=['total_bill', 'tip', 'size'])  # 仅选择 3 个变量做展示
g11.map_diag(sns.violinplot,
             inner='box')  # 对角线：小提琴图  # 同时展示核密度与“inner=box”的箱线细节
g11.map_offdiag(sns.scatterplot, alpha=0.5, s=40)  # 非对角线：散点图  # 半透明与较大点径提升可读性
g11.add_legend(title='性别')  # 添加图例并命名
g11.fig.suptitle(
    '示例13: 小提琴图组合\n(对角线=小提琴图)',  # 整体标题与说明
    fontsize=14,
    fontweight='bold',
    y=1.01)
plt.tight_layout()  # 布局优化
plt.show()  # 渲染显示：分组分布形态通过小提琴图更直观

# ========== 示例14: 热图显示相关性 ==========  # 示例14 小标题：自定义 2D 直方图热图函数
print("\n# ========== 示例14: 热图 + 散点图 ==========")  # 打印该示例的标题


def heatmap_func(x, y, **kwargs):  # 自定义函数：在单元格绘制微型 2D 直方图热图
    """在每个格子中绘制微型热图"""  # 文档字符串：说明用途
    ax = plt.gca()  # 获取当前轴对象
    # 创建2D直方图  # 将连续的 x、y 数据进行网格化统计，bins=10 表示 10×10 的格点
    hist, xedges, yedges = np.histogram2d(
        x, y, bins=10)  # hist 为计数矩阵；xedges/yedges 为边界
    ax.imshow(
        hist.T,  # 使用 imshow 显示转置后的计数矩阵（与坐标系匹配）
        origin='lower',  # 原点置于左下角，使坐标方向与数据更直观
        aspect='auto',  # 自适应长宽比，填满单元
        cmap='YlOrRd',  # 颜色映射：由浅黄到深红，计数越高颜色越深
        extent=[
            x.min(),
            x.max(),
            y.min(),  # extent 指定图像在数据坐标中的覆盖范围
            y.max()
        ])


g12 = sns.PairGrid(
    iris[['sepal_length', 'sepal_width', 'petal_length']],  # 仅选择 3 个变量做对比
    height=3)  # 子图高度更大，便于观察热度差异
g12.map_upper(sns.scatterplot, alpha=0.5, s=30)  # 上三角：半透明散点  # 直观显示点分布
g12.map_lower(heatmap_func)  # 下三角：热图  # 高密度区域呈现更深颜色，表达“点群密集度”
g12.map_diag(sns.histplot, kde=True)  # 对角线：直方 + KDE  # 同时展示频数与平滑分布
g12.fig.suptitle(
    '示例14: 热图展示密度\n(下三角=2D热图)',  # 整体标题与说明
    fontsize=14,
    fontweight='bold',
    y=1.01)
plt.tight_layout()  # 自动调整边距，避免标题/轴标签拥挤
plt.show()  # 渲染显示：上三角看散点形态，下三角看密度分布，互补展示
```

```python
# ==================== 第三部分：实际应用案例 ====================  # 章节标题注释：进入面向实战的 PairGrid 应用示例
print("\n" + "=" * 80)  # 打印 80 个“=”的分隔线；前置换行以提升可读性
print("【第三部分】实际应用案例")  # 打印当前部分标题
print("=" * 80)  # 再打印一次分隔线，形成视觉区块

# ========== 案例1: 数据探索性分析 ==========  # 案例1 小标题：以企鹅数据为例进行 EDA
print("\n# ========== 案例1: 企鹅数据探索性分析 ==========")  # 控制台输出本案例标题
g13 = sns.PairGrid(
    penguins,  # 创建 PairGrid：以 penguins 数据集为输入
    vars=[  # 指定要参与成对关系可视化的变量列
        'bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g'
    ],
    hue='species',  # 按企鹅品种着色分组
    height=2.5,  # 每个子图高度（英寸）
    palette='deep')  # 调色板：'deep'，区分度高
# 上三角：散点图（查看关系）  # 说明：用于展示两变量的相关形态与分布重叠
g13.map_upper(
    sns.scatterplot,
    s=40,  # 散点大小
    alpha=0.6,  # 透明度，缓解遮挡
    edgecolor='white',  # 白色描边，让点在深色背景更清晰
    linewidth=0.5)  # 描边线宽
# 下三角：KDE密度图（查看分布重叠）  # 说明：二维核密度等高线，聚类区域更明显
g13.map_lower(sns.kdeplot, levels=5, alpha=0.5)  # levels 指定等高线条数；半透明叠加更自然
# 对角线：KDE（查看单变量分布）  # 说明：单变量平滑分布曲线，观察峰态与离散
g13.map_diag(sns.kdeplot, linewidth=2.5, fill=True, alpha=0.6)  # 填充配合较粗线条提升辨识
g13.add_legend(title='企鹅品种', fontsize=10, title_fontsize=11)  # 添加图例并设置标题/字体

# 添加整体标题和说明  # 说明：suptitle 作为整幅图的总标题，包含布局提示
g13.fig.suptitle(
    '案例1: 企鹅体征数据探索\n(上=散点关系, 下=密度重叠, 对角线=分布)',
    fontsize=14,  # 标题字号
    fontweight='bold',  # 标题加粗
    y=1.01)  # 上移，避免与子图重叠
plt.tight_layout()  # 自动调整子图间距，减少元素重叠
plt.show()  # 渲染显示：得到按品种分组的成对关系矩阵

print("📊 分析发现:")  # 输出要点小结标题
print("  • Adelie企鹅喙较短但深度较大")  # 观察要点：品种间形态差异
print("  • Gentoo企鹅体型最大")  # 观察要点：体重/鳍长整体更大
print("  • Chinstrap企鹅喙最长")  # 观察要点：喙长维度突出

# ========== 案例2: 相关性分析 + 统计检验 ==========  # 案例2 小标题：结合相关系数与显著性标注
print("\n# ========== 案例2: 相关性分析与统计检验 ==========")  # 控制台输出本案例标题


def scatter_with_corr(x, y, **kwargs):  # 自定义单元格绘图：散点 + 相关/显著性 + 回归线
    """散点图 + 显示相关系数和显著性"""  # 文档字符串：说明功能
    ax = plt.gca()  # 获取当前坐标轴以进行绘图与标注
    ax.scatter(x, y, alpha=0.5, s=30, **kwargs)  # 绘制散点：半透明减少遮挡；s 控制点径
    # 计算统计量  # 使用皮尔逊相关检验衡量线性关系强度与显著性
    r, p = stats.pearsonr(x, y)  # r 为相关系数（-1~1），p 为显著性水平
    # 添加回归线  # 采用一阶多项式拟合（最小二乘）绘制趋势线
    z = np.polyfit(x, y, 1)  # 求解线性回归系数（斜率、截距）
    p_line = np.poly1d(z)  # 构造回归函数
    x_line = np.linspace(x.min(), x.max(), 100)  # 在 x 范围内生成等距点
    ax.plot(x_line, p_line(x_line), "r-", alpha=0.8, linewidth=2)  # 绘制红色回归线
    # 显示统计信息  # 依据 p 值映射显著性星级（***、**、*、ns）
    sig = "***" if p < 0.001 else "**" if p < 0.01 else "*" if p < 0.05 else "ns"
    ax.text(
        0.05,  # 文本框左下角 x（轴域比例坐标）
        0.95,  # 文本框左下角 y（轴域比例坐标）
        f'r={r:.3f}{sig}',  # 显示 r 值与显著性符号
        transform=ax.transAxes,  # 以轴域坐标放置文本，不随数据缩放
        fontsize=10,  # 字号
        va='top',  # 顶部对齐
        bbox=dict(boxstyle='round', facecolor='wheat',
                  alpha=0.7))  # 圆角背景框，提高可读性


g14 = sns.PairGrid(custom_data, vars=['变量A', '变量B', '变量C', '变量D'],
                   height=3)  # 构建 PairGrid：4 个自定义变量
g14.map_upper(scatter_with_corr)  # 上三角：散点+相关系数  # 使用自定义函数把可视与统计结合
g14.map_lower(sns.kdeplot, levels=5, color='blue',
              alpha=0.3)  # 下三角：密度图  # 低饱和蓝色，突出等高线
g14.map_diag(sns.histplot, kde=True, color='green',
             alpha=0.6)  # 对角线：分布  # 直方 + KDE，更直观
g14.fig.suptitle('案例2: 相关性分析\n(上=相关系数+显著性, 下=2D密度, ***p<0.001)',
                 fontsize=14,
                 fontweight='bold',
                 y=1.01)  # 抬高总标题
plt.tight_layout()  # 紧凑布局
plt.show()  # 渲染显示：便于解读变量相关与显著性水平

# ========== 案例3: 分组对比分析 ==========  # 案例3 小标题：构造 AB 测试数据并进行可视化对比
print("\n# ========== 案例3: 分组对比分析（AB测试可视化）==========")  # 控制台输出本案例标题

# 创建AB测试数据  # 说明：A/B 两组分别按不同均值/方差采样，模拟指标差异
np.random.seed(42)  # 固定随机种子，保证结果可复现
ab_test_data = pd.DataFrame({
    '转化率':
    np.concatenate([
        np.random.normal(0.05, 0.01, 100),  # A组  # 转化率均值低、方差小
        np.random.normal(0.07, 0.012, 100)
    ]),  # B组  # 转化率均值略高
    '客单价':
    np.concatenate(
        [np.random.normal(120, 20, 100),
         np.random.normal(130, 22, 100)]),  # B 组客单价略高
    '停留时间':
    np.concatenate(
        [np.random.normal(180, 40, 100),
         np.random.normal(210, 45, 100)]),  # B 组停留更久
    '分组': ['对照组A'] * 100 + ['实验组B'] * 100  # 标注所属组别
})  # 构造三指标 + 分组标签的数据框

g15 = sns.PairGrid(
    ab_test_data,  # 以合成数据为输入
    hue='分组',  # 按分组着色
    height=3.5,  # 子图尺寸更大以便观察
    palette={  # 自定义颜色映射，直观区分 AB
        '对照组A': 'blue',
        '实验组B': 'red'
    })
# 上三角：散点图  # 观察指标对之间的关系与组间分布差异
g15.map_upper(sns.scatterplot, alpha=0.5, s=50)  # 半透明 + 大点径提升可读性
# 下三角：KDE等高线  # 以等高线表现二维密度，强调聚集区域
g15.map_lower(sns.kdeplot, levels=5, linewidths=2)  # 更粗线条以增强辨识度
# 对角线：密度对比  # 单变量分布对比，便于观察组间偏移
g15.map_diag(sns.kdeplot, linewidth=3, fill=True, alpha=0.4)  # 填充配合较粗线形
g15.add_legend(title='测试分组',
               fontsize=11,
               title_fontsize=12,
               frameon=True,
               shadow=True)  # 带边框和阴影的图例，更突出

g15.fig.suptitle('案例3: AB测试效果对比\n(红色=实验组, 蓝色=对照组)',
                 fontsize=14,
                 fontweight='bold',
                 y=1.01)  # 上移避免重叠
plt.tight_layout()  # 自动布局
plt.show()  # 渲染显示：直观看到实验组整体指标上移

print("📊 AB测试结论:")  # 输出结论标题
print("  • 实验组转化率提升约40%")  # 样例性总结（基于设定的均值差异）
print("  • 客单价略有提升")  # 观察点
print("  • 用户停留时间增加")  # 观察点
print("  • 建议推广实验组方案")  # 行动建议

# ========== 案例4: 高维数据降维展示 ==========  # 案例4 小标题：构造多特征相关网络并用 PairGrid 辅助洞察
print("\n# ========== 案例4: 高维特征关系网络 ==========")  # 控制台输出本案例标题

# 创建具有复杂关系的数据  # 说明：先生成独立正态，再手工引入相关关系
np.random.seed(42)  # 固定随机性
n = 150  # 样本量
feature_data = pd.DataFrame({
    'F1': np.random.normal(0, 1, n),  # 特征 F1：标准正态
    'F2': np.random.normal(0, 1, n),  # 特征 F2：先独立，后面再覆盖成相关
    'F3': np.random.normal(0, 1, n),  # 特征 F3：先独立，后面再重写为相关组合
    'F4': np.random.normal(0, 1, n),  # 特征 F4：保持独立
    'F5': np.random.normal(0, 1, n),  # 特征 F5：先独立，后面再覆盖成相关
})  # 初始 DataFrame

# 添加一些关联  # 引入线性组合与噪声，模拟真实数据中的相关结构
feature_data['F2'] = feature_data['F1'] * 0.8 + np.random.normal(
    0, 0.3, n)  # F2 ~ 0.8*F1 + 噪声
feature_data[
    'F3'] = feature_data['F1'] * 0.5 + feature_data[  # F3 同时受 F1、F2 线性影响
        'F2'] * 0.4 + np.random.normal(0, 0.5, n)
feature_data['F4'] = np.random.normal(0, 1, n)  # 独立特征  # 保持为独立参照
feature_data['F5'] = feature_data['F3'] * 0.6 + np.random.normal(
    0, 0.4, n)  # F5 与 F3 强相关

# 添加聚类标签  # 使用 KMeans 聚 3 类用于上色分组
from sklearn.cluster import KMeans  # 导入 KMeans 聚类算法

kmeans = KMeans(n_clusters=3, random_state=42)  # 指定聚类数与随机种子
feature_data['聚类'] = kmeans.fit_predict(
    feature_data[['F1', 'F2', 'F3', 'F4', 'F5']])  # 对 5 维特征进行聚类，返回簇索引
feature_data['聚类'] = feature_data['聚类'].map({
    0: '簇A',
    1: '簇B',
    2: '簇C'
})  # 将簇索引映射为标签名

g16 = sns.PairGrid(feature_data, hue='聚类', height=2,
                   palette='Set1')  # 构建 PairGrid：按簇着色
g16.map_upper(sns.scatterplot, s=20, alpha=0.5)  # 上三角：散点图  # 小点径 + 半透明以减少遮挡
g16.map_lower(corrfunc)  # 显示相关系数  # 下三角调用前文定义的 corrfunc 展示 r 与 p
g16.map_diag(sns.histplot, kde=True, alpha=0.7)  # 对角线：直方 + KDE  # 单变量分布与密度
g16.add_legend(title='聚类结果', fontsize=9)  # 添加图例并设置字体

g16.fig.suptitle('案例4: 特征工程 - 特征关系网络\n(下三角=相关系数, 发现强相关特征)',
                 fontsize=14,
                 fontweight='bold',
                 y=1.01)  # 上移标题以避免重叠
plt.tight_layout()  # 紧凑布局
plt.show()  # 渲染显示：结合聚类与相关，洞察冗余与结构

print("📊 特征分析:")  # 输出要点标题
print("  • F1、F2、F3、F5 高度相关，考虑降维")  # 结论：可考虑 PCA 等方法降维
print("  • F4 独立，保留")  # 结论：F4 提供互补信息
print("  • 数据自然分成3个簇")  # 结论：聚类结构清晰

# ==================== 第四部分：高级技巧 ====================  # 章节标题注释：总结若干提升可视化质量的技巧
print("\n" + "=" * 80)  # 分隔条，与上文风格一致
print("【第四部分】高级技巧与最佳实践")  # 打印本部分标题
print("=" * 80)  # 结束分隔线

# ========== 技巧1: 动态调整子图大小 ==========  # 技巧1 小标题：使用 corner 只展示下三角，减少冗余
print("\n# ========== 技巧1: corner参数 - 只显示下三角 ==========")  # 控制台输出本技巧标题
g17 = sns.PairGrid(iris, hue='species', corner=True,
                   height=2.5)  # corner=True 省略上三角，减少重复信息
g17.map_lower(sns.scatterplot, alpha=0.6, s=40)  # 下三角：散点图  # 半透明 + 中等点径
g17.map_diag(sns.histplot, kde=True, alpha=0.7)  # 对角线：直方 + KDE  # 展示单变量分布
g17.add_legend()  # 添加图例
g17.fig.suptitle('技巧1: corner=True\n(避免重复，只显示下三角)',
                 fontsize=14,
                 fontweight='bold',
                 y=1.01)  # 抬高标题
plt.tight_layout()  # 紧凑布局
plt.show()  # 渲染显示：更聚焦、更节省空间

# ========== 技巧2: 精细控制图例 ==========  # 技巧2 小标题：演示图例位置/样式的精细化配置
print("\n# ========== 技巧2: 精细控制图例 ==========")  # 控制台输出本技巧标题
g18 = sns.PairGrid(tips, hue='day', height=2.5,
                   palette='tab10')  # 使用 tips 数据；按星期几着色
g18.map_diag(sns.kdeplot, linewidth=2.5, fill=True,
             alpha=0.5)  # 对角线：KDE 填充，较粗线条
g18.map_offdiag(sns.scatterplot, alpha=0.5, s=30)  # 非对角线：散点图  # 观察变量关系及分组差异

# 自定义图例  # 说明：将图例置于右侧，带阴影/圆角/标题字体控制
g18.add_legend(
    title='星期几',
    bbox_to_anchor=(1.05, 0.5),  # 将图例锚到右侧外部
    loc='center left',  # 图例定位
    frameon=True,  # 带边框
    shadow=True,  # 阴影效果
    fancybox=True,  # 圆角边框
    title_fontsize=12,  # 标题字号
    fontsize=10)  # 条目字号

g18.fig.suptitle('技巧2: 自定义图例\n(位置、边框、阴影)',
                 fontsize=14,
                 fontweight='bold',
                 y=1.01)  # 抬高标题
plt.tight_layout()  # 自动留白，确保右侧图例不遮挡
plt.show()  # 渲染显示：图例样式更美观、信息更清晰

print("\n# ========== 技巧3: 渐变调色板 ==========")  # 技巧3 小标题：按点密度进行颜色渐变编码

# 1. 只选需要的列，并 drop 掉任何含 NaN 的行  # 数据清洗：避免 NaN 影响 KDE/散点着色
penguins_subset = penguins[[
    'bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g'
]].dropna()  # 仅保留完整样本行

g19 = sns.PairGrid(penguins_subset, height=2.5,
                   diag_sharey=False)  # 构建网格；对角线独立 Y 轴范围


def scatter_colored(x, y, **kwargs):  # 自定义函数：按局部点密度上色的散点
    ax = plt.gca()  # 获取当前轴对象
    # 2. 再次确保无 NaN（防御式）  # 对传入的一对变量进行二次清理
    mask = ~(np.isnan(x) | np.isnan(y))
    x, y = x[mask], y[mask]
    if len(x) < 2:  # 样本太少无法估计密度  # 少于 2 个点直接散点展示
        ax.scatter(x, y, color='grey', s=30, alpha=0.3)
        return
    xy = np.vstack([x, y])  # 组合为 2×N 数组供 KDE 使用
    z = stats.gaussian_kde(xy)(xy)  # 估计每个点的密度值 z（越大表示越密集）
    idx = z.argsort()  # 根据密度从低到高排序，先绘低密度再绘高密度避免遮挡
    ax.scatter(
        x.iloc[idx],
        y.iloc[idx],
        c=z[idx],  # 以密度值作为颜色映射输入
        s=30,  # 点径
        cmap='viridis',  # 渐变色盘：低密到高密循序变深
        alpha=0.6)  # 半透明提升叠加可读性


g19.map_upper(scatter_colored)  # 上三角：按密度着色散点  # 高密区域颜色更深
g19.map_lower(sns.kdeplot, cmap='viridis', fill=True,
              alpha=0.5)  # 下三角：与上三角配套的密度填色
g19.map_diag(sns.histplot, kde=True, color='steelblue',
             alpha=0.7)  # 对角线：直方 + KDE（钢蓝配色）

g19.fig.suptitle('技巧3: 按密度着色\n(高密度区域颜色更深)',
                 fontsize=14,
                 fontweight='bold',
                 y=1.01)  # 上移总标题
plt.tight_layout()  # 紧凑布局，避免文字/图例重叠
plt.show()  # 渲染显示：实现“密度即颜色强度”的视觉编码
```
