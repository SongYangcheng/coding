关于KNN手写体识别，精度和准确率都比不上后期的深度学习模型，但是KNN算法简单易懂，适合初学者入门机器学习分类任务。下面是一个使用KNN算法对钻石价格进行分类的完整示例代码。


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn import datasets
from sklearn.model_selection import train_test_split
import matplotlib
digits = datasets.load_digits()
X = digits.data
y = digits.target
y.shape
```

```python
digits.target_names
```

```python
some_digit = X[220] #选择第一个样本
some_digit_image = some_digit.reshape(8, 8) #将一维数组重塑为8x8的二维数组
plt.imshow(some_digit_image,  #显示图像
           cmap=matplotlib.cm.binary, #使用黑白色图像
           interpolation="nearest") #插值方法为最近邻插值
plt.show()
```

```python
#数据分类

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
X_train.shape, X_test.shape
```

```python
#模型训练
from sklearn.neighbors import KNeighborsClassifier
knn_clf = KNeighborsClassifier(n_neighbors=3) #创建KNN分类器实例，设置邻居数为3
knn_clf.fit(X_train, y_train) #使用训练数据训练模型

```

```python
#计算准确率
y_predict = knn_clf.predict(X_test)
from sklearn.metrics import accuracy_score
#accuracy_score是计算分类准确率的函数= 正确预测的数量/总预测数量
accuracy_score(y_test, y_predict)
```
